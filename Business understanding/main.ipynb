شرکت می‌خواهد پیش از ازدست‌دادن مشتریان، افرادی را که در معرض ریزش قرار دارند شناسایی کند و برای هر گروه، راهبرد مناسبی جهت حفظ مشتری پیشنهاد دهد.

اما یک مشکل وجود دارد: دیتاست تراکنش‌ها مستقیماً ستون «ریزش مشتری» ندارد. علاوه بر این، اطلاعات شامل لغو سفارش، مرجوعی، مقادیر منفی، شناسه‌های گمشده، مشتریان با رفتارهای بسیار متفاوت و داده‌های زمانی است.

مأموریت ما: از داده‌های خام تراکنش‌ها یک سامانه تحلیلی بسازید که ریسک ریزش مشتری را پیش‌بینی کند، عوامل مؤثر را تحلیل کند و برای اجرای یک کمپین حفظ مشتری پیشنهاد عملی ارائه دهد.

 هدف: با استفاده از سابقه خرید مشتریان، مدلی بسازیم که پیش‌بینی کنه کدوم مشتری‌ها در آینده

خروجی شامل دیتاست آماده‌شده، ویژگی‌های استخراج‌شده، مدل‌های آزمایش‌شده، نتایج ارزیابی و فهرستی رتبه‌بندی‌شده از مشتریان براساس ریسک پیش‌بینی‌شده ریزش است. یکی از خروجی‌های عملی پروژه، معرفی ۵۰۰ مشتری با بالاترین ریسک است؛ مشروط بر اینکه تعداد مشتریان واجد شرایط کافی باشد.

###  سوال 24: چگونه از نتایج مدل برای بهبود کسب‌وکار استفاده می‌کنیم؟

فروشگاه می‌تواند مشتریان پرریسک را در اولویت بررسی و کمپین‌های حفظ مشتری قرار دهد و براساس ارزش خرید و رفتار هر گروه، اقدام مناسب‌تری انتخاب کند. سپس با استفاده از آزمایش‌هایی مانند آ/ب تِستینگ نتایج گروه دریافت‌کننده کمپین را با یک گروه کنترل مقایسه می‌کنیم تا مشخص شود اقدامات انجام‌شده واقعاً به افزایش خرید مجدد و کاهش ریزش منجر شده‌اند یا خیر.

# جمع‌بندی پروژه 

در این پروژه، با تحلیل سابقه خرید مشتریان و ساخت یک مدل دسته‌بندی، مشتریان در معرض ریزش پیش‌بینی و براساس ریسک اولویت‌بندی می‌شوند. هدف نهایی تنها ساخت یک مدل با معیارهای فنی مناسب نیست، بلکه فراهم‌کردن اطلاعاتی کاربردی برای تصمیم‌گیری بهتر در بازاریابی و حفظ مشتری است. میزان اثرگذاری واقعی این تصمیم‌ها باید پس از اجرای اقدامات مناسب در محیط کسب‌وکار ارزیابی شود.
