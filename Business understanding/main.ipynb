# برنامه‌ریزی پروژه داده‌کاوی: پیش‌بینی ریزش مشتریان

## 🔹 بخش اول: تعیین اهداف کسب‌وکار (Determine Business Objectives)

### ۱. مسئله اصلی کسب‌وکار چیست، چرا حل آن اهمیت دارد و در صورت اجرا نشدن پروژه چه پیامدهایی خواهد داشت؟

**پاسخ:**

یک فروشگاه آنلاین بین‌المللی متوجه شده است که برخی از مشتریان قدیمی به‌تدریج خریدهای خود را کاهش می‌دهند یا ناگهان خرید را متوقف می‌کنند. مشکل اصلی این است که شرکت معمولاً زمانی متوجه از دست دادن مشتری می‌شود که دیگر فرصت مناسبی برای حفظ او وجود ندارد. از طرفی، دیتاست موجود نیز ستونی با عنوان «ریزش مشتری» ندارد.

در صورت حل نشدن این مسئله:
- مشتریان ارزشمند بدون شناسایی قبلی از دست می‌روند.
- بودجه بازاریابی بدون هدف‌گیری مناسب هزینه می‌شود.
- جذب مشتریان جدید هزینه بیشتری نسبت به حفظ مشتریان فعلی خواهد داشت.
- شرکت دید روشنی از میزان ریسک ریزش مشتریان خود نخواهد داشت.

**چرا این سؤال پرسیده شد؟**

برای مشخص شدن مسئله واقعی کسب‌وکار، ایجاد توافق میان مدیران و اعضای تیم و توجیه ارزش اقتصادی اجرای پروژه.

---

### ۲. ذی‌نفعان اصلی پروژه چه کسانی هستند و هرکدام چه نیازها و انتظاراتی از نتایج پروژه دارند؟

**پاسخ:**

| ذی‌نفع | نیاز اصلی | انتظار از پروژه |
|---|---|---|
| مدیریت ارشد | حفظ درآمد و رشد پایدار | گزارش ساده، تصمیم قابل‌دفاع و بازگشت سرمایه روشن |
| بازاریابی و CRM | شناسایی مشتریان در معرض ریزش | لیست رتبه‌بندی‌شده ۵۰۰ مشتری و پیشنهاد اقدام برای هر گروه |
| واحد مالی | کنترل هزینه کمپین | تحلیل نقطه سربه‌سر و سناریوهای مالی شفاف |
| فروش و پشتیبانی | شناخت مشتریان مشکل‌دار | درک الگوهای مرجوعی و لغو سفارش |
| تیم IT و داده | داده تمیز و قابل‌بازتولید | مستندات و کد قابل‌اجرا |
| مشتریان | تجربه خرید مناسب | پیشنهادهای مرتبط، بدون مزاحمت یا ارسال پیام‌های ناخواسته |

**چرا این سؤال پرسیده شد؟**

برای شناسایی نیازهای مختلف ذی‌نفعان و جلوگیری از تعارض میان آن‌ها؛ برای مثال، واحد بازاریابی ممکن است خواهان تماس بیشتر با مشتریان باشد، درحالی‌که واحد مالی به دنبال کاهش هزینه‌ها است.

---

### ۳. هدف اصلی و اهداف فرعی کسب‌وکار چیست و چه شاخص‌هایی برای سنجش موفقیت آن‌ها در نظر گرفته می‌شود؟

**پاسخ:**

**هدف اصلی کسب‌وکار:**

«کاهش ریزش مشتریان ارزشمند با شناسایی زودهنگام آن‌ها و اجرای کمپین حفظ مشتری به‌صورت هدفمند.»

**اهداف فرعی:**

۱. ساخت برچسب ریزش مشتری بر اساس رفتار خرید.

۲. رتبه‌بندی مشتریان بر اساس میزان ریسک ریزش.

۳. شناسایی عوامل مؤثر بر ریزش مشتریان، مانند فاصله زمانی از آخرین خرید.

۴. دسته‌بندی مشتریان بر اساس رفتار خرید برای طراحی اقدامات مناسب هر گروه.

**شاخص‌های کلیدی عملکرد (KPI):**

| شاخص | تعریف | نوع |
|---|---|---|
| Precision@500 | چه تعداد از ۵۰۰ مشتری انتخاب‌شده واقعاً در معرض ریزش هستند؟ | فنی و کسب‌وکاری |
| Recall | چه درصدی از کل مشتریان ریزشی شناسایی شده‌اند؟ | فنی |
| PR-AUC | کیفیت رتبه‌بندی مدل در شرایط نامتوازن بودن داده‌ها | فنی |
| Lift | مدل چند برابر بهتر از انتخاب تصادفی مشتریان عمل می‌کند؟ | فنی و کسب‌وکاری |
| سود خالص مورد انتظار کمپین | ارزش مشتریان حفظ‌شده منهای هزینه اجرای کمپین | مالی |
| نرخ سربه‌سر | حداقل نرخ موفقیت لازم برای جلوگیری از ضرر کمپین | مالی |

**معیار موفقیت از دیدگاه مدیران:**

اگر کمپین ۵۰۰ نفره سود خالص مثبتی ایجاد کند و نرخ ریزش مشتریان ارزشمند کاهش یابد، پروژه از دید مدیران موفق خواهد بود. مدیران بیشتر به نتایج مالی و تصمیم‌های قابل‌اجرا اهمیت می‌دهند تا جزئیات فنی مدل.

**معیار موفقیت از دیدگاه بازاریابی و CRM:**

لیست ۵۰۰ نفره باید قابل‌فهم و قابل‌اجرا باشد، درصد بالایی از مشتریان انتخاب‌شده واقعاً در معرض ریزش باشند و برای هر گروه مشتری اقدام مشخصی پیشنهاد شود.

**چرا این سؤال پرسیده شد؟**

برای تبدیل هدف کلی کسب‌وکار به اهداف قابل‌اندازه‌گیری و تعیین معیارهایی که بتوان با استفاده از آن‌ها موفقیت واقعی پروژه را ارزیابی کرد.

---

### ۴. مزایای مالی، عملیاتی و مرتبط با تجربه مشتری که از اجرای پروژه انتظار می‌رود چیست؟

**پاسخ:**

| نوع مزیت | دستاوردهای مورد انتظار |
|---|---|
| مالی | حفظ درآمد حاصل از مشتریان ارزشمند، کاهش هزینه تماس‌های بی‌اثر و افزایش بازده سرمایه‌گذاری بازاریابی (ROI) |
| عملیاتی | اولویت‌بندی خودکار مشتریان، تصمیم‌گیری مبتنی بر داده به‌جای حدس و اجرای کمپین‌های قابل‌تکرار |
| تجربه مشتری | ارائه پیشنهادهای مرتبط‌تر، یادآوری به‌موقع خرید و کاهش پیام‌های تبلیغاتی نامرتبط |

**چرا این سؤال پرسیده شد؟**

برای مشخص شدن ارزش پروژه از جنبه‌های مختلف و اطمینان از اینکه خروجی آن فقط یک مدل پیش‌بینی نیست، بلکه می‌تواند به بهبود عملکرد واقعی کسب‌وکار کمک کند.

---

### ۵. اگر مدل از نظر فنی عملکرد خوبی داشته باشد اما نتواند اعتماد و رضایت تیم بازاریابی را جلب کند، آیا پروژه موفق محسوب می‌شود؟

**پاسخ:**

خیر. مدلی که توسط کاربران نهایی مورد استفاده قرار نگیرد، ارزش عملیاتی و اقتصادی ایجاد نمی‌کند.

همچنین Accuracy به‌تنهایی معیار مناسبی برای ارزیابی این پروژه نیست؛ زیرا در صورت نامتوازن بودن کلاس‌ها، حتی مدلی که تمام مشتریان را غیرریزش‌کننده پیش‌بینی کند، ممکن است Accuracy بالایی داشته باشد.

به همین دلیل باید از معیارهایی مانند Precision، Recall، PR-AUC و قابلیت تفسیر مدل استفاده کنیم.

**چرا این سؤال پرسیده شد؟**

برای تأکید بر اینکه موفقیت پروژه فقط به عملکرد فنی مدل وابسته نیست و پذیرش خروجی توسط کاربران نهایی نیز اهمیت دارد.

---

### ۶. هدف رسمی کسب‌وکار (Business Objective Statement) چیست؟

**پاسخ:**

«شرکت می‌خواهد با تحلیل تراکنش‌های ۱۲ ماه گذشته، مشتریانی را که در ۹۰ روز آینده خریدی نخواهند داشت، پیش از وقوع ریزش شناسایی کند و با استفاده از بودجه تماس با ۵۰۰ مشتری، سود خالص کمپین حفظ مشتری را به حداکثر برساند.»

**چرا این سؤال پرسیده شد؟**

برای ایجاد یک تعریف رسمی و مشخص از هدف کسب‌وکار که تمام مراحل بعدی پروژه باید با آن هماهنگ باشند.

---

## 🔹 بخش دوم: ارزیابی وضعیت موجود (Assess Situation)

### ۷. پیش از شروع پروژه چه اطلاعاتی باید از شرکت دریافت کنیم و برای شناخت بهتر مسئله چه پرسش‌هایی باید از مدیران و تیم بازاریابی مطرح شود؟

**پاسخ:**

پیش از شروع پروژه باید اطلاعات زیر از شرکت دریافت شود:

- تعریف شرکت از مشتری فعال و مشتری ریزش‌کرده.
- حاشیه سود محصولات و سفارش‌ها.
- هزینه هر تماس و تخفیف ارائه‌شده.
- سیاست‌های مربوط به مرجوعی و لغو سفارش.
- دسترسی‌های مجاز به اطلاعات مشتریان.

**پرسش‌های مهم از مدیر شرکت:**

۱. از دید شما مشتری ریزش‌کرده چه کسی است؟ آیا ۹۰ روز بدون خرید، معیار مناسبی برای تشخیص ریزش محسوب می‌شود؟

۲. حفظ کدام گروه از مشتریان برای شرکت اهمیت بیشتری دارد؟

۳. بودجه کمپین و هزینه هر تماس با مشتری چقدر است؟

۴. حاشیه سود متوسط هر سفارش چقدر است؟

۵. حداقل بازده قابل‌قبول کمپین (ROI) چقدر است؟

۶. آیا تاکنون کمپین حفظ مشتری اجرا شده است؟ نتایج آن چه بوده است؟

۷. آیا امکان اجرای آزمایش A/B (A/B Test) برای بررسی اثربخشی کمپین وجود دارد؟

۸. کدام کشورها یا بازارها برای شرکت اولویت بیشتری دارند؟

۹. مدل پیش‌بینی باید هر چند وقت یک‌بار به‌روزرسانی شود؟

۱۰. چه محدودیت‌های قانونی و حریم خصوصی برای برقراری ارتباط با مشتریان وجود دارد؟

**پرسش‌های مهم از تیم بازاریابی و CRM:**

۱. کدام اقدام مانند تخفیف، یادآوری خرید یا پیشنهاد محصول تاکنون نتیجه بهتری داشته است؟

۲. در هر دوره چه تعداد مشتری را می‌توانید عملاً تحت پوشش کمپین قرار دهید؟

۳. چه فرمتی از لیست مشتریان برای شما کاربردی‌تر است؟

۴. چه مشتریانی نباید مورد تماس قرار بگیرند؛ مانند مشتریان مسدودشده یا دارای شکایت؟

۵. آیا رفتار مشتریان عمده‌فروش و خرده‌فروش با یکدیگر متفاوت است؟

۶. مرجوعی و لغو سفارش معمولاً به چه دلایلی اتفاق می‌افتد؟

۷. فصل‌ها و مناسبت‌ها چه تأثیری بر رفتار خرید مشتریان دارند؟

۸. چه عواملی باعث می‌شود به نتایج و پیش‌بینی‌های مدل اعتماد کنید؟

۹. آیا اطلاعات تماس مشتریان مانند ایمیل و شماره تلفن در دسترس است؟

۱۰. نتایج هر کمپین چگونه ثبت و پیگیری می‌شود؟

**چرا این سؤال پرسیده شد؟**

برای اینکه تعریف ریزش، محدودیت‌های مالی، نیازهای عملیاتی و شکل خروجی پروژه بر اساس اطلاعات واقعی شرکت تعیین شوند، نه صرفاً فرضیات تحلیل‌گر.

---

### ۸. چه محدودیت‌هایی از نظر مالی، فنی، قانونی و اطلاعاتی در اجرای پروژه وجود دارد و این محدودیت‌ها چه تأثیری بر روند کار خواهند داشت؟

**پاسخ:**

| نوع محدودیت | شرح محدودیت | تأثیر بر پروژه |
|---|---|---|
| مالی | بودجه شرکت فقط امکان تماس با ۵۰۰ مشتری را فراهم می‌کند. | مدل باید دقیق‌ترین ۵۰۰ مشتری را از نظر ریسک ریزش شناسایی کند. |
| فنی | دیتاست فاقد ستون مستقیم ریزش مشتری است و اطلاعاتی مانند سن و میزان رضایت مشتری نیز وجود ندارد. | باید برچسب ریزش و ویژگی‌های موردنیاز از داده‌های موجود ساخته شوند. |
| فنی | بخشی از رکوردها فاقد CustomerID هستند. | این رکوردها برای تحلیل رفتار مشتری در سطح فردی قابل‌استفاده نیستند. |
| فنی | داده‌ها فقط مربوط به یک سال، از دسامبر ۲۰۱۰ تا دسامبر ۲۰۱۱ هستند. | تشخیص الگوهای فصلی و روندهای بلندمدت محدود خواهد بود. |
| قانونی | اطلاعات مشتریان مشمول مقررات حریم خصوصی مانند GDPR هستند. | شناسه‌های مشتریان باید ناشناس‌سازی شوند و دسترسی به اطلاعات محدود باشد. |
| علّی | دیتاست به‌تنهایی تأثیر تخفیف یا تماس با مشتری را اثبات نمی‌کند. | نتایج اقتصادی باید با فرضیات شفاف گزارش شوند و اثبات اثر واقعی کمپین به A/B Test نیاز دارد. |

**چرا این سؤال پرسیده شد؟**

برای تعیین مرزهای واقعی پروژه، جلوگیری از وعده‌های غیرقابل‌اجرا و مشخص شدن محدودیت‌هایی که باید در تحلیل و گزارش نهایی در نظر گرفته شوند.

---

### ۹. مهم‌ترین ریسک‌های فنی، تحلیلی و اجرایی پروژه چیست و برای مدیریت هرکدام چه راهکاری در نظر گرفته می‌شود؟

**پاسخ:**

| ریسک | احتمال وقوع | شدت اثر | راهکار مدیریت |
|---|---|---|---|
| نشت اطلاعات آینده به ویژگی‌های مدل | متوسط | بسیار زیاد | استفاده از اطلاعات گذشته برای ساخت ویژگی‌ها و ارزیابی زمانی |
| تشخیص اشتباه مشتریان فصلی به‌عنوان مشتری ریزشی با معیار ۹۰ روز | زیاد | زیاد | بررسی تعریف ریزش و انجام تحلیل حساسیت |
| نامتوازن بودن کلاس‌ها | متوسط | متوسط | استفاده از Precision، Recall و PR-AUC به‌جای اتکا به Accuracy |
| کیفیت پایین داده‌ها به دلیل لغو سفارش، مقادیر منفی و داده‌های گمشده | زیاد | متوسط | ممیزی داده و مستندسازی تصمیم‌های مربوط به حذف یا نگهداری رکوردها |
| برداشت اشتباه از اهمیت ویژگی‌ها به‌عنوان رابطه علت و معلولی | زیاد | متوسط | تأکید بر تفاوت همبستگی و علیت |
| پیچیده بودن مدل و دشواری درک نتایج توسط تیم بازاریابی | متوسط | متوسط | استفاده از مدل پایه ساده و تفسیرپذیر در کنار مدل‌های پیچیده‌تر |

**چرا این سؤال پرسیده شد؟**

برای شناسایی مشکلات احتمالی پیش از وقوع آن‌ها و طراحی راهکارهایی که احتمال شکست پروژه را کاهش دهند.

---

### ۱۰. چه داده‌هایی برای اجرای پروژه موردنیاز هستند، کدام اطلاعات ممکن است در دسترس نباشند و در صورت کمبود یا وجود خطا در داده‌ها چه باید کرد؟

**پاسخ:**

**داده‌های موردنیاز:**

- شماره فاکتور (InvoiceNo)
- شناسه محصول (StockCode)
- تعداد کالا (Quantity)
- تاریخ تراکنش (InvoiceDate)
- قیمت محصول (UnitPrice)
- شناسه مشتری (CustomerID)
- کشور مشتری (Country)

**داده‌های سخت‌دسترس یا ناموجود:**

- دلیل دقیق مرجوعی محصولات
- کانال خرید مشتری
- اطلاعات جمعیت‌شناختی مشتریان
- سوابق ارتباطات و تماس‌های قبلی با مشتری

در صورت کافی نبودن اطلاعات، می‌توان از ویژگی‌هایی که مستقیماً از تراکنش‌ها استخراج می‌شوند، مانند شاخص‌های RFM و روند تغییر رفتار خرید استفاده کرد. همچنین باید محدودیت‌های ناشی از نبود اطلاعات را در گزارش نهایی شفاف بیان کنیم و به شرکت پیشنهاد دهیم اطلاعات بیشتری را در آینده ثبت کند.

در صورت وجود خطا در داده‌ها نیز باید پیش از هرگونه حذف یا اصلاح، ماهیت رکوردها بررسی شود. برای مثال، اگر سفارش‌های لغوشده یا مقادیر منفی بدون بررسی حذف یا با سایر تراکنش‌ها جمع شوند، محاسبه ارزش خرید و فاصله زمانی از آخرین خرید (Recency) اشتباه خواهد شد و حتی برچسب ریزش مشتری نیز ممکن است نادرست ساخته شود.

**چرا این سؤال پرسیده شد؟**

برای مشخص شدن قابلیت‌های واقعی دیتاست، شناسایی اطلاعات موردنیاز و جلوگیری از تصمیم‌های اشتباه هنگام پاک‌سازی و آماده‌سازی داده‌ها.

---

### ۱۱. وضعیت کلی پروژه از نظر نقاط قوت، نقاط ضعف، فرصت‌ها و تهدیدها چگونه ارزیابی می‌شود؟ (SWOT Analysis)

**پاسخ:**

| عامل | موارد شناسایی‌شده |
|---|---|
| **نقاط قوت (Strengths)** | وجود داده‌های تراکنشی واقعی و حجیم، هدف مشخص پروژه و امکان استخراج ویژگی‌های رفتاری مشتریان |
| **نقاط ضعف (Weaknesses)** | نبود برچسب آماده ریزش، نبود اطلاعات جمعیت‌شناختی و محدود بودن داده‌ها به یک سال |
| **فرصت‌ها (Opportunities)** | کاهش هزینه جذب مشتریان جدید، بهبود بازده بازاریابی و امکان اجرای کمپین‌های حفظ مشتری به‌صورت مستمر |
| **تهدیدها (Threats)** | نشت اطلاعات، تفسیر اشتباه روابط علّی، بی‌اعتمادی تیم بازاریابی به مدل و محدودیت‌های قانونی حریم خصوصی |

**چرا این سؤال پرسیده شد؟**

برای داشتن دیدی جامع نسبت به شرایط داخلی و خارجی پروژه و شناسایی عواملی که می‌توانند بر موفقیت یا شکست آن تأثیر بگذارند.

---

## 🔹 بخش سوم: تعیین اهداف داده‌کاوی (Determine Data Mining Goals)

### ۱۲. هدف اصلی داده‌کاوی در این پروژه چیست و چه تفاوتی با هدف کسب‌وکار دارد؟

**پاسخ:**

هدف اصلی داده‌کاوی، ساخت مدلی است که با استفاده از رفتار خرید مشتریان در ۱۸۰ روز گذشته، احتمال نداشتن خرید در ۹۰ روز آینده را برای هر مشتری برآورد کند.

تفاوت این دو هدف به شکل زیر است:

| نوع هدف | توضیح |
|---|---|
| هدف کسب‌وکار (Business Goal) | مشخص می‌کند شرکت چه چیزی می‌خواهد؛ یعنی کاهش ریزش مشتریان و افزایش سود. |
| هدف داده‌کاوی (Data Mining Goal) | مشخص می‌کند چگونه با استفاده از داده‌ها به هدف کسب‌وکار می‌رسیم؛ یعنی پیش‌بینی ریسک ریزش و دسته‌بندی رفتاری مشتریان. |

**هدف رسمی داده‌کاوی:**

«پیش‌بینی ریسک ریزش مشتریان با استفاده از ویژگی‌های رفتاری ۱۸۰ روز گذشته و گروه‌بندی آن‌ها بر اساس الگوی رفتار خرید.»

**چرا این سؤال پرسیده شد؟**

برای جلوگیری از اشتباه گرفتن اهداف کسب‌وکار با اهداف فنی و مشخص شدن نقش تحلیل داده و مدل‌سازی در حل مسئله اصلی.

---

### ۱۳. مسئله داده‌کاوی پروژه از چه نوعی است و چرا از این روش‌ها استفاده می‌کنیم؟

**پاسخ:**

این پروژه از دو روش اصلی داده‌کاوی استفاده می‌کند:

| نوع مسئله | نقش در پروژه | دلیل انتخاب |
|---|---|---|
| طبقه‌بندی (Classification) | هدف اصلی | پیش‌بینی اینکه مشتری در آینده ریزش می‌کند یا خیر. برچسب این مسئله باید از رفتار خرید ساخته شود. |
| خوشه‌بندی (Clustering) | هدف مکمل | تقسیم مشتریان به گروه‌های رفتاری مختلف برای طراحی اقدامات مناسب هر گروه. |
| رگرسیون (Regression) | هدف اصلی نیست | هدف فعلی پیش‌بینی یک مقدار عددی نیست؛ اما اگر بخواهیم ارزش خرید آینده مشتری را پیش‌بینی کنیم، می‌توان از رگرسیون استفاده کرد. |
| قوانین انجمنی (Association Rules) | در اولویت این پروژه نیست | مسئله اصلی ما کشف کالاهای هم‌خرید یا الگوهای سبد خرید نیست. |

اگر خروجی موردنظر تغییر کند، نوع مسئله نیز ممکن است تغییر کند:

- احتمال ریزش مشتری ← طبقه‌بندی (Classification)
- ارزش یا میزان خرید آینده ← رگرسیون (Regression)
- گروه‌بندی مشتریان بر اساس رفتار ← خوشه‌بندی (Clustering)

**چرا این سؤال پرسیده شد؟**

برای انتخاب صحیح روش‌ها و الگوریتم‌های موردنیاز پروژه. در این سناریو طبقه‌بندی هدف اصلی و خوشه‌بندی یک روش مکمل است.

---

### ۱۴. برای ارزیابی عملکرد مدل چه معیارهایی باید استفاده شوند، هرکدام چه مفهومی دارند و کدام‌یک در این پروژه اهمیت بیشتری دارد؟

**پاسخ:**

معیارهای مورد استفاده در این پروژه عبارت‌اند از:

| معیار | مفهوم |
|---|---|
| Accuracy | درصد پیش‌بینی‌های صحیح از کل پیش‌بینی‌ها |
| Precision | از میان مشتریانی که مدل ریزشی تشخیص داده، چه تعداد واقعاً ریزشی بوده‌اند؟ |
| Recall | از میان تمام مشتریان ریزشی واقعی، چه تعداد توسط مدل شناسایی شده‌اند؟ |
| F1-Score | میانگین هماهنگ Precision و Recall |
| PR-AUC | میزان کیفیت عملکرد مدل در ایجاد تعادل میان Precision و Recall |
| ROC-AUC | توانایی مدل در تفکیک مشتریان ریزشی از غیرریزش‌کننده |
| Confusion Matrix | نمایش تعداد پیش‌بینی‌های درست و اشتباه مدل در هر کلاس |
| Precision@500 | درصد مشتریان واقعاً ریزشی در میان ۵۰۰ مشتری اولویت‌بندی‌شده |

**آیا Accuracy به‌تنهایی کافی است؟**

خیر. به دلیل احتمال نامتوازن بودن تعداد مشتریان ریزشی و غیرریزش‌کننده، Accuracy ممکن است تصویر اشتباهی از عملکرد مدل ارائه دهد.

**کدام معیار اهمیت بیشتری دارد؟**

با توجه به اینکه بودجه شرکت فقط امکان تماس با ۵۰۰ مشتری را فراهم می‌کند، معیار Precision@500 اهمیت بیشتری دارد. زیرا هدف این است که از میان ۵۰۰ مشتری انتخاب‌شده، بیشترین تعداد ممکن واقعاً در معرض ریزش باشند.

در عین حال، Recall نیز گزارش می‌شود تا مشخص شود چه سهمی از کل مشتریان ریزشی شناسایی شده‌اند.

**چرا این سؤال پرسیده شد؟**

برای انتخاب معیارهای ارزیابی متناسب با مسئله و جلوگیری از انتخاب مدلی که از نظر عددی عملکرد خوبی دارد اما از نظر کسب‌وکار کاربردی نیست.

---

### ۱۵. خطاهای مدل چه هزینه‌هایی برای کسب‌وکار ایجاد می‌کنند و کدام نوع خطا اهمیت بیشتری دارد؟

**پاسخ:**

دو نوع خطای اصلی در این پروژه وجود دارد:

**خطای مثبت کاذب (False Positive):**

مشتری در واقع ریزش نمی‌کند، اما مدل او را در معرض ریزش تشخیص می‌دهد.

پیامدهای احتمالی:
- هزینه تماس غیرضروری با مشتری
- ارائه تخفیف بدون نیاز
- اشغال یکی از ظرفیت‌های محدود ۵۰۰ تماس

**خطای منفی کاذب (False Negative):**

مشتری در واقع در معرض ریزش قرار دارد، اما مدل موفق به شناسایی او نمی‌شود.

پیامدهای احتمالی:
- از دست رفتن مشتری
- از دست رفتن درآمد احتمالی آینده
- از دست رفتن فرصت حفظ مشتری

به‌طور کلی، اگر ارزش مشتری از هزینه تماس بیشتر باشد، خطای منفی کاذب (FN) می‌تواند خسارت بیشتری ایجاد کند. بااین‌حال، به دلیل محدودیت بودجه، خطای مثبت کاذب (FP) نیز مستقیماً بر سودآوری کمپین تأثیر می‌گذارد.

بنابراین، اهمیت هر خطا باید با توجه به هزینه تماس، هزینه تخفیف و ارزش اقتصادی مشتری بررسی شود.

**چرا این سؤال پرسیده شد؟**

برای مشخص شدن پیامدهای اقتصادی اشتباهات مدل و استفاده از این اطلاعات در انتخاب مدل نهایی و تحلیل سودآوری کمپین.

---

### ۱۶. موفقیت نهایی مدل چگونه تعریف می‌شود و چه معیارهای فنی و کسب‌وکاری برای پذیرش آن در نظر گرفته می‌شود؟

**پاسخ:**

مدل زمانی موفق محسوب می‌شود که عملکرد آن از خط مبنای ساده مبتنی بر RFM بهتر باشد، بتواند مشتریان پرریسک را به‌درستی رتبه‌بندی کند و لیست ۵۰۰ نفره آن از نظر اقتصادی ظرفیت ایجاد سود مثبت داشته باشد.

**معیارهای فنی موفقیت:**

۱. عملکرد PR-AUC بهتر از خط مبنا باشد.

۲. مقدار Precision@500 از نرخ شناسایی تصادفی مشتریان ریزشی بیشتر باشد؛ یعنی Lift بزرگ‌تر از ۱ باشد.

۳. عملکرد مدل در تاریخ‌های مرجع مختلف پایدار باشد.

**معیارهای کسب‌وکاری موفقیت:**

۱. سود خالص مورد انتظار کمپین مثبت باشد.

۲. نرخ سربه‌سر کمپین واقع‌بینانه باشد.

۳. برای هر گروه رفتاری مشتریان، اقدام مشخص و قابل‌اجرا تعریف شده باشد.

**هدف رسمی داده‌کاوی (Data Mining Goal Statement):**

«ساخت یک مدل طبقه‌بندی که با استفاده از ویژگی‌های رفتاری ۱۸۰ روز قبل از تاریخ مرجع، مشتریانی را که در ۹۰ روز آینده هیچ خرید معتبری نخواهند داشت، رتبه‌بندی کند. همچنین با استفاده از خوشه‌بندی رفتاری، مشتریان به گروه‌های مختلف تقسیم شوند تا اقدامات مناسب برای هر گروه طراحی شود. تمام ارزیابی‌ها باید به‌صورت زمانی و بدون نشت اطلاعات انجام شوند.»

**ارتباط هدف داده‌کاوی با هدف کسب‌وکار:**

مدل، میزان ریسک ریزش مشتری را مشخص می‌کند؛ خوشه‌بندی، نوع رفتار مشتری را نشان می‌دهد و کمپین حفظ مشتری، اقدام مناسب را تعیین می‌کند. ترکیب این سه بخش به تحقق هدف اصلی کسب‌وکار کمک می‌کند.

**چرا این سؤال پرسیده شد؟**

برای مشخص شدن معیار پذیرش مدل و اطمینان از اینکه خروجی فنی پروژه مستقیماً به تصمیم‌ها و اهداف واقعی کسب‌وکار متصل است.
