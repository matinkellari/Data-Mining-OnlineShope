# فاز ۱ CRISP-DM: Business Understanding
## پروژه: شکار مشتریان در آستانه ریزش

---

## بخش ۱: Determine Business Objectives

**سوال 1:** مسئله اصلی کسب‌وکار را با زبان ساده توضیح دهید.

**پاسخ:** یک فروشگاه آنلاین بین‌المللی می‌بیند بعضی مشتریان قدیمی کم‌کم کمتر خرید میکنند و بعضی بی‌هشدار خرید را متوقف می‌کنند. شرکت معمولاً بعد از رفتن مشتری متوجه می‌شود و می‌خواهد مشتریان در معرض ریزش را پیش از رفتن پیدا کند.

**چرایی پرسش:** برای فهمیدن اینکه آیا میتوانیم مسئله واقعی کسب و کار را تشخیص بدیم

---

**سوال 2:** اگر پروژه اجرا نشود چه مشکلاتی ایجاد می‌شود؟

**پاسخ:** مشتریان ارزشمند بی‌صدا از دست می‌روند، بودجه بازاریابی بدون هدف خرج می‌شود و... ، جذب مشتری جدید هم معمولاً گران‌تر از حفظ مشتری فعلی است.

**چرایی پرسش:** یکی از بهترین روش‌ها برای تعیین اولویت حل پروژه، بررسی پیامدهای عدم اجرای پروژه است. و این سوال هزینه «انجام‌ندادن» پروژه را نشان می‌دهد و توجیه‌کننده آن است

---

**سوال 3:** چه ذی‌نفعانی (Stakeholders) در پروژه حضور دارند؟

**پاسخ:** مدیریت ارشد، واحد بازاریابی/CRM، واحد مالی، واحد فروش و پشتیبانی، تیم IT/داده و مشتریان نهایی.

**چرایی پرسش:** اگر حتی یکی از ذی‌نفعان یا همان کسانی که سود میبرند نادیده گرفته شود، ممکن است پروژه با مشکلاتی مثل:

عدم همکاری کاربران ، مقاومت در برابر تغییر، کمبود بودجه، عدم دسترسی به داده‌ها، مشکلات قانونی، و... مواجه شود.



---

**سوال 4:** نیاز و انتظار هر ذی‌نفع چیست؟

**پاسخ:**

| ذی‌نفع | نیاز | انتظار از پروژه |
|--------|------|-----------------|
| مدیریت ارشد | حفظ درآمد و رشد پایدار | گزارش ساده و تصمیم قابل‌دفاع با بازگشت سرمایه روشن |
| بازاریابی/CRM | لیست مشتریان هدف برای تماس | لیست رتبه‌بندی‌شده و پیشنهاد اقدام برای هر گروه |
| مالی | کنترل هزینه کمپین | تحلیل سربه‌سر و سناریوهای مالی شفاف |
| فروش/پشتیبانی | شناخت مشتریان مسئله‌دار | درک الگوی لغو و مرجوعی |
| IT/داده | داده قابل‌اعتماد و کد قابل‌بازتولید | مستندات قابل‌اجرا |
| مشتری | تجربه خرید خوب | پیشنهاد مرتبط، بدون مزاحمت |

**چرایی پرسش:** ممکن است با وجود کیفیت بالای مدل، پروژه شکست بخورد، چون انتظارات ذینغعان شناسایی و مدیریت نشده است.

---

**سوال 5:** هدف اصلی کسب‌وکار را در یک جمله بنویسید.

**پاسخ:** کاهش ریزش مشتریان ارزشمند با شناسایی زودهنگام آن‌ها و اجرای یک کمپین حفظ هدفمند.

**چرایی پرسش:** همه تصمیم‌های بعدی باید با این جمله هم‌جهت باشند و هدفمون مشخص باشه.

---

**سوال 6:** حداقل سه هدف فرعی تعریف کنید.

**پاسخ:** (1) ساخت برچسب ریزش از رفتار خرید. (2) رتبه‌بندی مشتریان بر اساس ریسک ریزش. (3) کشف عوامل مرتبط با ریزش. (4) گروه‌بندی رفتاری مشتریان برای طراحی اقدام هر گروه.

**چرایی پرسش:** تقسیم هدف اصلی به اهداف فرعی باعث می‌شود:

پروژه قابل برنامه‌ریزی‌تر شود.
شاخص‌های عملکرد (KPIs) راحت‌تر تعریف شوند.
مسئولیت‌ها بین اعضای تیم تقسیم شوند.
پیشرفت پروژه قابل ارزیابی باشد.

---

**سوال 7:** چه شاخص‌هایی نشان می‌دهد پروژه موفق بوده است؟

**پاسخ:** هم شاخص‌های کسب‌وکاری (سود خالص کمپین، کاهش ریزش مشتریان ارزشمند) و هم شاخص‌های فنی (کیفیت رتبه‌بندی مدل نسبت به روش ساده). هر دو لازم‌اند، چون مدل خوب بدون سود واقعی موفقیت نیست.

> ۱. افزایش نرخ شناسایی مشتریان در معرض ریزش

اگر سیستم بتواند درصد بیشتری از مشتریانی را که واقعاً در آستانه ریزش هستند شناسایی کند، پروژه از نظر شناسایی به‌موقع مشتریان موفق بوده است.


> ۲. کاهش تعداد مشتریان ازدست‌رفته

اگر پس از اجرای راهکارهای پیشنهادی، تعداد مشتریانی که خرید خود را متوقف می‌کنند کاهش پیدا کند، نشان‌دهنده اثربخشی پروژه در حفظ مشتریان است.


> ۳. افزایش نرخ بازگشت مشتریان غیرفعال

اگر مشتریانی که مدتی خریدی انجام نداده‌اند، پس از اجرای کمپین‌های پیشنهادی دوباره خرید کنند، پروژه در بازگرداندن مشتریان موفق بوده است.


> ۴. افزایش سود خالص کمپین حفظ مشتری

اگر درآمد حاصل از بازگشت و حفظ مشتریان، پس از کسر هزینه‌های تخفیف، تبلیغات و اجرای کمپین بیشتر از هزینه‌ها باشد، پروژه از نظر اقتصادی موفق بوده است.


> ۵. افزایش دقت شناسایی مشتریان پرریسک

اگر از میان مشتریانی که مدل به‌عنوان مشتریان در معرض ریزش معرفی می‌کند، تعداد بیشتری واقعاً در معرض ریزش باشند، منابع شرکت هدفمندتر مصرف شده‌اند.


> ۶. بهبود بازگشت سرمایه (ROI)

اگر میزان سود حاصل از اجرای راهکارهای حفظ مشتری نسبت به هزینه‌های انجام‌شده افزایش پیدا کند، نشان‌دهنده بازده اقتصادی مناسب پروژه است

---

**سوال 8:** حداقل پنج KPI تعریف کنید.


### شاخص‌های ارزیابی عملکرد مدل (KPI)

| KPI | تعریف | هدف نمونه (فرضی) |
|---|---|---|
| **Precision** | از میان مشتریانی که مدل ریزشی تشخیص داده، چه تعداد واقعاً ریزش کرده‌اند؟ | بالاتر از ۷۰٪ |
| **Recall** | مدل چه درصدی از کل مشتریان ریزشی را شناسایی کرده است؟ | بالاتر از ۷۵٪ |
| **F1-score** | میانگین هماهنگ Precision و Recall که تعادل بین این دو معیار را نشان می‌دهد. | بالاتر از ۷۲٪ |
| **PR-AUC** | توانایی مدل در ایجاد تعادل مناسب بین Precision و Recall در آستانه‌های مختلف پیش‌بینی را نشان می‌دهد. | بالاتر از ۰٫۷۰ |
| **ROC-AUC** | توانایی مدل در تفکیک مشتریان ریزشی از مشتریان غیرریزش‌کننده را اندازه‌گیری می‌کند. | بالاتر از ۰٫۸۰ |
| **Confusion Matrix** | تعداد پیش‌بینی‌های درست و اشتباه مدل را در چهار گروه مثبت واقعی، منفی واقعی، مثبت کاذب و منفی کاذب نمایش می‌دهد. | کاهش مثبت و منفی کاذب |


---

**سوال 9:** موفقیت پروژه از دید مدیر چگونه اندازه‌گیری می‌شود؟

**پاسخ:** اگر کمپین ۵۰۰ نفره سود خالص مثبت داشته باشد و تصمیم قابل‌اجرا و شفافی ارائه شود. مدیر به دقت مدل کاری ندارد، به پول و ریسک تصمیم نگاه می‌کند.

**چرایی پرسش:** مدیران سازمان‌ها معمولاً سؤالات متفاوتی می‌پرسند:

آیا پروژه باعث کاهش ریزش شد؟
آیا سرمایه‌گذاری انجام‌شده ارزش داشت؟ و...
 تعریف موفقیت از دید تصمیم‌گیرنده اصلی.

---

**سوال 10:** موفقیت پروژه از دید تیم بازاریابی (کاربر عملیاتی) چگونه اندازه‌گیری می‌شود؟

**پاسخ:** لیست ۵۰۰ نفره قابل‌فهم و قابل‌اجرا باشد، درصد بالایی از آن واقعاً مشتریان در خطر باشند و برای هر گروه اقدام مشخص وجود داشته باشد.

**چرایی پرسش:** این تیم مصرف‌کننده مستقیم خروجی است و اگر استفاده نکند پروژه بی‌اثر می‌ماند.

---

**سوال 11:** اگر سیستم ۹۰٪ دقت داشته باشد اما تیم بازاریابی به آن اعتماد نکند، پروژه موفق است؟ چرا؟

**پاسخ:** خیر. مدلی که استفاده نشود سود نمی‌سازد. ضمناً Accuracy گمراه‌کننده است: در داده نامتوازن، مدلی که همه را «غیرریزشی» بگوید هم دقت بالا می‌گیرد.

**چرایی پرسش:** نشان می‌دهد چرا به جای Accuracy از Precision، Recall، PR-AUC و تفسیرپذیری استفاده می‌کنیم.

---

**سوال 12:** مزایای مالی احتمالی پروژه را فهرست کنید.

**پاسخ:** حفظ درآمد مشتریان ارزشمند، کاهش هزینه تماس‌های بی‌اثر و افزایش بازده سرمایه‌گذاری بازاریابی (ROI).

**چرایی پرسش:** توجیه اقتصادی پروژه برای مدیر و مالی.

---

**سوال 13:** مزایای عملیاتی احتمالی پروژه را فهرست کنید.

**پاسخ:** اولویت‌بندی خودکار مشتریان، تصمیم‌گیری مبتنی بر داده به جای حدس، و امکان تکرار کمپین به‌صورت دوره‌ای.

**چرایی پرسش:** نشان می‌دهد پروژه یک‌بارمصرف نیست و در فرآیند کاری جا می‌افتد.

---

**سوال 14:** مزایای احتمالی برای تجربه مشتری را فهرست کنید.

**پاسخ:** پیشنهاد مرتبط‌تر، یادآوری به‌موقع خرید و کاهش پیام‌های بی‌ربط به مشتری.

**چرایی پرسش:** اثر پروژه فقط مالی نیست و رضایت مشتری را هم شامل می‌شود.

---

**سوال 15:** یک Business Objective Statement رسمی بنویسید.

**پاسخ:** «شرکت می‌خواهد با تحلیل تراکنش‌های گذشته، مشتریانی را که در ۹۰ روز آینده خرید معتبری نخواهند داشت پیش از ریزش شناسایی کند و با بودجه تماس برای ۵۰۰ مشتری، سود خالص کمپین حفظ را به حداکثر برساند.»

**چرایی پرسش:** سندی رسمی که کل پروژه به آن ارجاع می‌دهد.

---

## بخش ۲: Assess Situation

**سوال 16:** چه اطلاعاتی را قبل از شروع پروژه باید از شرکت دریافت کنیم؟

**پاسخ:** تعریف شرکت از «مشتری فعال و ریزش‌کرده»، حاشیه سود، هزینه هر تماس و تخفیف، سیاست مرجوعی و لغو، دسترسی‌های مجاز به داده و دوره زمانی داده‌های موجود.

**چرایی پرسش:** بدون این‌ها تعریف ریزش و تحلیل اقتصادی حدسی می‌شود.

---

**سوال 17:** حداقل ۱۰ سوال که باید از مدیر شرکت پرسیده شود.

**پاسخ:**

1. از دید شما «مشتری ریزش‌کرده» کیست؟ ۹۰ روز بدون خرید منطقی است؟
2. حفظ کدام گروه از مشتریان مهم‌تر است (پرهزینه یا پرتعداد)؟
3. بودجه کمپین و هزینه هر تماس چقدر است؟
4. حاشیه سود متوسط هر سفارش چقدر است؟
5. حداقل بازده قابل‌قبول کمپین چقدر است؟
6. آیا قبلاً کمپین حفظ مشتری اجرا شده و نتیجه‌اش چه بوده؟
7. آیا امکان اجرای A/B Test وجود دارد؟
8. کدام بازارها اولویت دارند؟
9. مدل هر چند وقت یک‌بار باید به‌روز شود؟
10. محدودیت‌های قانونی ارتباط با مشتری چیست؟

**چرایی پرسش:** پاسخ‌ها فرض‌ها را به واقعیت تبدیل می‌کند (تعریف ریزش، بودجه، سنجش اثر).

---

**سوال 18:** حداقل ۱۰ سوال که باید از تیم بازاریابی/CRM پرسیده شود.

**پاسخ:**

1. کدام اقدام (تخفیف، یادآوری، پیشنهاد محصول) تاکنون بهتر جواب داده؟
2. در هر دوره عملاً چند مشتری را می‌توانید پوشش دهید؟
3. لیست در چه قالبی برای شما قابل‌استفاده‌تر است؟
4. کدام مشتریان نباید تماس گرفته شوند؟
5. مشتریان عمده‌فروش و خرده‌فروش چه تفاوتی دارند؟
6. لغو و مرجوعی معمولاً چه دلایلی دارد؟
7. فصل‌ها و مناسبت‌ها چه اثری بر خرید دارند؟
8. چه چیزی باعث می‌شود به خروجی مدل اعتماد کنید؟
9. اطلاعات تماس مشتریان در دسترس است؟
10. نتیجه هر کمپین چگونه ثبت و پیگیری می‌شود؟

**چرایی پرسش:** قالب خروجی و اقدام‌های پیشنهادی باید با واقعیت کار تیم بازاریابی جور باشد.

---

**سوال 19:** محدودیت‌های مالی پروژه را تحلیل کنید.

**پاسخ:** بودجه تماس فقط برای ۵۰۰ مشتری است، پس مدل باید دقیق‌ترین ۵۰۰ نفر را انتخاب کند. هر تماس بی‌نتیجه هم هزینه است و هم یک جای خالی از ۵۰۰ جا.

**چرایی پرسش:** بودجه محدود مهم‌ترین عامل انتخاب معیار ارزیابی است (Precision در ۵۰۰ نفر اول).

---

**سوال 20:** محدودیت‌های فنی پروژه را تحلیل کنید.

**پاسخ:** برچسب ریزش آماده نیست و باید ساخته شود. فقط داده تراکنش داریم (بدون اطلاعات جمعیتی یا رضایت)، داده‌ها زمانی و متوالی‌اند و احتمال نقص داده (شناسه گمشده، لغو، مقادیر منفی) وجود دارد.

**چرایی پرسش:** محدودیت‌های فنی روش ساخت برچسب، ویژگی‌ها و ارزیابی را تعیین می‌کنند.

---

**سوال 21:** محدودیت‌های قانونی و حریم خصوصی را تحلیل کنید.

**پاسخ:** مشتریان بین‌المللی‌اند و قوانین حفاظت از داده (مانند GDPR) اعمال می‌شود. شناسه‌ها باید در خروجی ناشناس یا مستعار باشند. همچنین داده این پروژه اثر علّی تخفیف یا تماس را اثبات نمی‌کند و ادعای اثر کمپین بدون A/B Test نادرست است.

**چرایی پرسش:** خروجی و ادعاهای پروژه باید قانونی و قابل‌دفاع بمانند.

---

**سوال 22:** مهم‌ترین ریسک‌ها را شناسایی و احتمال و شدت اثرشان را مشخص کنید.

**پاسخ:**

| ریسک | احتمال | شدت اثر | اقدام مقابله |
|------|--------|---------|--------------|
| نشت اطلاعات آینده به ویژگی‌ها | متوسط | بسیار زیاد | ویژگی فقط از گذشته، ارزیابی زمانی |
| تعریف ۹۰ روزه، مشتری فصلی را ریزشی نشان دهد | زیاد | زیاد | تحلیل حساسیت آستانه |
| عدم‌توازن کلاس‌ها | متوسط | متوسط | معیارهای Precision/Recall/PR-AUC |
| کیفیت پایین داده (لغو، منفی، گمشده) | زیاد | متوسط | ممیزی مستند، بدون حذف کورکورانه |
| تفسیر علّی اشتباه از اهمیت ویژگی | زیاد | متوسط | تأکید بر همبستگی، نه علیت |
| مدل پیچیده و غیرقابل‌فهم برای بازاریابی | متوسط | متوسط | مدل ساده‌ی قابل‌توضیح کنار مدل پیچیده |

**چرایی پرسش:** ریسک شناخته‌شده قابل مدیریت است.

---

**سوال 23:** چه داده‌هایی احتمالاً برای پروژه لازم است؟

**پاسخ:** تراکنش‌های خرید شامل شماره فاکتور، محصول، تعداد، تاریخ، قیمت واحد، شناسه مشتری و کشور، برای حداقل چند ماه متوالی.

**چرایی پرسش:** تعیین می‌کند چه ویژگی‌هایی قابل‌ساخت است.

---

**سوال 24:** جمع‌آوری کدام داده‌ها ممکن است دشوار باشد؟

**سوال 25:** اگر داده کافی نباشد چه راهکارهایی پیشنهاد می‌کنید؟

**چرایی پرسش:** آمادگی برای کمبود داده، بدون وعده‌های غیرواقعی.

**سوال 26:** اگر داده‌ها خطا داشته باشند چه مشکلاتی ایجاد می‌شود؟

**پاسخ:** لغو یا مقدار منفی اگر به‌اشتباه جمع یا حذف شود، ارزش خرید و فاصله از آخرین خرید را غلط می‌کند. این باعث برچسب ریزش مخدوش و مدل گمراه می‌شود.

**چرایی پرسش:** پروژه صراحتاً حذف مقادیر منفی بدون بررسی معنا را مجاز نمی‌داند.

## بخش ۳: Determine Data Mining Goals

**سوال 28:** هدف داده‌کاوی این پروژه چیست؟

**پاسخ:** ساخت مدلی که با رفتار خرید ۱۸۰ روز گذشته هر مشتری، احتمال نداشتن خرید معتبر در ۹۰ روز آینده را برآورد کند و مشتریان را بر اساس رفتار گروه‌بندی کند.

**چرایی پرسش:** ترجمه نیاز کسب‌وکار به زبان فنی.

---

**سوال 29:** تفاوت Business Goal و Data Mining Goal چیست؟

**پاسخ:** هدف کسب‌وکار می‌گوید «چه می‌خواهیم» (کاهش ریزش و سود). هدف داده‌کاوی می‌گوید «با داده چطور به آن برسیم» (پیش‌بینی ریسک و خوشه‌بندی).

**چرایی پرسش:** اشتباه رایج، یکی گرفتن این دو است.

---

**سوال 30:** هدف داده‌کاوی را در یک جمله بنویسید.

**پاسخ:** پیش‌بینی ریسک ریزش مشتریان با ویژگی‌های رفتاری گذشته و گروه‌بندی رفتاری آن‌ها، با ارزیابی زمانی و بدون نشت اطلاعات.

**چرایی پرسش:** جمله‌ای دقیق و قابل‌سنجش برای راهنمایی فاز Modeling.

---

**سوال 31:** پروژه Classification، Regression، Clustering یا Association Rule است؟

**پاسخ:**

| نوع | وضعیت | دلیل |
|-----|--------|------|
| Classification | ✅ هدف اصلی | خروجی دو‌حالته: ریزش می‌کند یا نه |
| Clustering | ✅ هدف مکمل | شناخت گروه‌های رفتاری و تعیین اقدام هر گروه |
| Regression | ❌ | هدف ما مقدار پیوسته نیست (مگر بخواهیم ارزش آینده را پیش‌بینی کنیم) |
| Association Rule | ❌ | هدف ما سبد کالا یا قوانین هم‌خریدی نیست |

**چرایی پرسش:** نوع مسئله، الگوریتم و معیار را تعیین می‌کند و بحث کلاس هم همین بود. پاسخ: هر دو، با نقش‌های متفاوت.

---

**سوال 32:** اگر خروجی «احتمال ریزش»، «میزان خرید آینده» یا «گروه‌بندی مشتریان» باشد، نوع مسئله چیست؟

**پاسخ:** احتمال ریزش: Classification (احتمالاتی). میزان خرید آینده: Regression. گروه‌بندی مشتریان: Clustering.

**چرایی پرسش:** نشان می‌دهد تغییر شکل خروجی، نوع مسئله را عوض می‌کند.

---

**سوال 33:** چه معیارهایی برای ارزیابی مدل مهم است؟

**پاسخ:** Precision، Recall، F1، PR-AUC، ROC-AUC، Confusion Matrix و Precision@500.

**چرایی پرسش:** معیار اشتباه، مدل اشتباه را «بهترین» نشان می‌دهد.

---

**سوال 34:** Accuracy چیست و آیا کافی است؟

**پاسخ:** درصد پیش‌بینی‌های درست از کل. کافی نیست، چون در داده نامتوازن مدل بی‌فایده هم دقت بالا می‌گیرد.

**چرایی پرسش:** پروژه صراحتاً Accuracy را معیار انتخاب مدل نهایی نمی‌داند.

---

**سوال 35:** Precision چیست؟

**پاسخ:** از مشتریانی که «ریزشی» پیش‌بینی کردیم، چند درصد واقعاً ریزش کردند.

**چرایی پرسش:** نشان می‌دهد بودجه تماس چقدر درست خرج شده است.

---

**سوال 36:** Recall چیست؟

**پاسخ:** از کل مشتریان واقعاً ریزش‌کننده، چند درصد را پیدا کردیم.

**چرایی پرسش:** نشان می‌دهد چه سهمی از ریزش‌ها را از دست داده‌ایم.

---

**سوال 37:** در این پروژه Precision مهم‌تر است یا Recall؟ چرا؟

**پاسخ:** چون فقط ۵۰۰ تماس داریم، Precision در ۵۰۰ نفر اول مهم‌تر است. هر مورد اشتباه یک جای خالی را هدر می‌دهد. Recall هم برای دیدن سهم پوشش گزارش می‌شود.

**چرایی پرسش:** محدودیت بودجه اولویت معیار را تعیین می‌کند.

---

**سوال 38:** هزینه False Positive چیست؟

**پاسخ:** مشتری سالمی که تماس یا تخفیف می‌گیرد: هزینه تماس، تخفیف بی‌مورد و یک جای خالی از ۵۰۰ جا.

**چرایی پرسش:** در تحلیل اقتصادی کمپین استفاده می‌شود.

---

**سوال 39:** هزینه False Negative چیست؟

**پاسخ:** مشتری در خطری که شناسایی نشد و رفت: از دست رفتن ارزش خرید آینده او.

**چرایی پرسش:** هزینه فرصت از دست رفته را نشان می‌دهد.

---

**سوال 40:** کدام خطا برای شرکت خطرناک‌تر است؟

**پاسخ:** معمولاً FN، اگر ارزش مشتری از هزینه تماس بیشتر باشد. ولی چون ظرفیت تماس محدود است، FP هم مستقیماً سود را می‌خورد. پس هر دو را در تحلیل حساسیت بررسی می‌کنیم.

**چرایی پرسش:** پاسخ به اعداد هزینه و ارزش مشتری وابسته است، نه شهود.

---

**سوال 41:** موفقیت مدل را چگونه تعریف می‌کنید؟

**پاسخ:** مدل موفق است که از خط مبنای RFM بهتر باشد، لیست ۵۰۰ نفره‌اش سود مثبت بسازد و ارزیابی زمانی بدون نشت داشته باشد.

**چرایی پرسش:** برنده و بازنده بودن مدل از قبل مشخص باشد.

---

**سوال 42:** سه معیار فنی برای موفقیت مدل تعریف کنید.

**پاسخ:** (1) PR-AUC بهتر از خط مبنای RFM. (2) Lift@500 بزرگ‌تر از 1. (3) پایداری عملکرد در تاریخ‌های مرجع مختلف.

**چرایی پرسش:** پل میان کیفیت فنی مدل و تصمیم کسب‌وکاری.

---

**سوال 43:** سه معیار کسب‌وکاری برای موفقیت پروژه تعریف کنید.

**پاسخ:** (1) سود خالص مورد انتظار مثبت. (2) نرخ سربه‌سر واقع‌بینانه. (3) اقدام مشخص و قابل‌اجرا برای هر گروه مشتری.

**چرایی پرسش:** مدل خوب بدون تصمیم سودآور موفقیت نیست.

**سوال 44:** یک Data Mining Goal Statement رسمی بنویسید.

**پاسخ:** «ساخت مدل طبقه‌بندی که با ویژگی‌های رفتاری ۱۸۰ روز قبل از تاریخ مرجع، مشتریانی را که در ۹۰ روز بعد خرید معتبری نخواهند داشت رتبه‌بندی کند، همراه با خوشه‌بندی رفتاری برای طراحی اقدام هر گروه، با ارزیابی زمانی و بدون نشت اطلاعات.»
