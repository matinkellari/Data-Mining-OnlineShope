# فاز ۱ CRISP-DM: Business Understanding
## پروژه: شکار مشتریان در آستانه ریزش

---

## بخش ۱: Determine Business Objectives

**سوال 1:** مسئله اصلی کسب‌وکار را با زبان ساده توضیح دهید.

**پاسخ:** یک فروشگاه آنلاین بین‌المللی می‌بیند بعضی مشتریان قدیمی کم‌کم کمتر خرید میکنند و بعضی بی‌هشدار خرید را متوقف می‌کنند. شرکت معمولاً بعد از رفتن مشتری متوجه می‌شود و می‌خواهد مشتریان در معرض ریزش را پیش از رفتن پیدا کند.

**چرایی پرسش:** برای فهمیدن اینکه آیا میتوانیم مسئله واقعی کسب و کار را تشخیص بدیم، در بیشتر سازمان‌ها، مدیر هنگام درخواست پروژه‌ داده‌کاوی معمولاً راه‌حل را مطرح می‌کند، نه مسئله را.

---

**سوال 2:** اگر پروژه اجرا نشود چه مشکلاتی ایجاد می‌شود؟

**پاسخ:** مشتریان ارزشمند بی‌صدا از دست می‌روند، بودجه بازاریابی بدون هدف خرج می‌شود و... ، جذب مشتری جدید هم معمولاً گران‌تر از حفظ مشتری فعلی است.

**چرایی پرسش:** یکی از بهترین روش‌ها برای تعیین اولویت حل پروژه، بررسی پیامدهای عدم اجرای پروژه است. و این سوال هزینه «انجام‌ندادن» پروژه را نشان می‌دهد و توجیه‌کننده آن است

---

**سوال 3:** چه ذی‌نفعانی (Stakeholders) در پروژه حضور دارند؟

**پاسخ:** مدیریت ارشد، واحد بازاریابی/CRM، واحد مالی، واحد فروش و پشتیبانی، تیم IT/داده و مشتریان نهایی.

**چرایی پرسش:** اگر حتی یکی از ذی‌نفعان یا همان کسانی که سود میبرند نادیده گرفته شود، ممکن است پروژه با مشکلاتی مثل:

عدم همکاری کاربران ، مقاومت در برابر تغییر، کمبود بودجه، عدم دسترسی به داده‌ها، مشکلات قانونی، و... مواجه شود.



---

**سوال 4:** نیاز و انتظار هر ذی‌نفع چیست؟

**پاسخ:**

| ذی‌نفع | نیاز | انتظار از پروژه |
|--------|------|-----------------|
| مدیریت ارشد | حفظ درآمد و رشد پایدار | گزارش ساده و تصمیم قابل‌دفاع با بازگشت سرمایه روشن |
| بازاریابی/CRM | لیست مشتریان هدف برای تماس | لیست رتبه‌بندی‌شده و پیشنهاد اقدام برای هر گروه |
| مالی | کنترل هزینه کمپین | تحلیل سربه‌سر و سناریوهای مالی شفاف |
| فروش/پشتیبانی | شناخت مشتریان مسئله‌دار | درک الگوی لغو و مرجوعی |
| IT/داده | داده قابل‌اعتماد و کد قابل‌بازتولید | مستندات قابل‌اجرا |
| مشتری | تجربه خرید خوب | پیشنهاد مرتبط، بدون مزاحمت |

**چرایی پرسش:** ممکن است با وجود کیفیت بالای مدل، پروژه شکست بخورد، چون انتظارات ذینغعان شناسایی و مدیریت نشده است.

---

**سوال 5:** هدف اصلی کسب‌وکار را در یک جمله بنویسید.

**پاسخ:** کاهش ریزش مشتریان ارزشمند با شناسایی زودهنگام آن‌ها و اجرای یک کمپین حفظ هدفمند.

**چرایی پرسش:** همه تصمیم‌های بعدی باید با این جمله هم‌جهت باشند و هدفمون مشخص باشه.

---

**سوال 6:** حداقل سه هدف فرعی تعریف کنید.

**پاسخ:** 	 (1) خوشه‌بندی مشتریان بر اساس رفتار، (2) ساخت برچسب ریزش، (3) پیش‌بینی ریسک ریزش با کشف عوامل.

**چرایی پرسش:** تقسیم هدف اصلی به اهداف فرعی باعث می‌شود:

پروژه قابل برنامه‌ریزی‌تر شود.
شاخص‌های عملکرد (KPIs) راحت‌تر تعریف شوند.
مسئولیت‌ها بین اعضای تیم تقسیم شوند.
پیشرفت پروژه قابل ارزیابی باشد.

---

**سوال 7:** چه شاخص‌هایی نشان می‌دهد پروژه موفق بوده است؟

**پاسخ:** هم شاخص‌های کسب‌وکاری (سود خالص کمپین، کاهش ریزش مشتریان ارزشمند) و هم شاخص‌های فنی (کیفیت رتبه‌بندی مدل نسبت به روش ساده). هر دو لازم‌اند، چون مدل خوب بدون سود واقعی موفقیت نیست.

> ۱. افزایش نرخ شناسایی مشتریان در معرض ریزش

اگر سیستم بتواند درصد بیشتری از مشتریانی را که واقعاً در آستانه ریزش هستند شناسایی کند، پروژه از نظر شناسایی به‌موقع مشتریان موفق بوده است.


> ۲. کاهش تعداد مشتریان ازدست‌رفته

اگر پس از اجرای راهکارهای پیشنهادی، تعداد مشتریانی که خرید خود را متوقف می‌کنند کاهش پیدا کند، نشان‌دهنده اثربخشی پروژه در حفظ مشتریان است.


> ۳. افزایش نرخ بازگشت مشتریان غیرفعال

اگر مشتریانی که مدتی خریدی انجام نداده‌اند، پس از اجرای کمپین‌های پیشنهادی دوباره خرید کنند، پروژه در بازگرداندن مشتریان موفق بوده است.


> ۴. افزایش سود خالص کمپین حفظ مشتری

اگر درآمد حاصل از بازگشت و حفظ مشتریان، پس از کسر هزینه‌های تخفیف، تبلیغات و اجرای کمپین بیشتر از هزینه‌ها باشد، پروژه از نظر اقتصادی موفق بوده است.


> ۵. افزایش دقت شناسایی مشتریان پرریسک

اگر از میان مشتریانی که مدل به‌عنوان مشتریان در معرض ریزش معرفی می‌کند، تعداد بیشتری واقعاً در معرض ریزش باشند، منابع شرکت هدفمندتر مصرف شده‌اند.


> ۶. بهبود بازگشت سرمایه (ROI)

اگر میزان سود حاصل از اجرای راهکارهای حفظ مشتری نسبت به هزینه‌های انجام‌شده افزایش پیدا کند، نشان‌دهنده بازده اقتصادی مناسب پروژه است

---

**سوال 8:** حداقل پنج KPI تعریف کنید.


### شاخص‌های ارزیابی عملکرد مدل (KPI)

| KPI | تعریف | هدف نمونه (فرضی) |
|---|---|---|
| **Precision** | از میان مشتریانی که مدل ریزشی تشخیص داده، چه تعداد واقعاً ریزش کرده‌اند؟ | بالاتر از ۷۰٪ |
| **Recall** | مدل چه درصدی از کل مشتریان ریزشی را شناسایی کرده است؟ | بالاتر از ۷۵٪ |
| **F1-score** | میانگین هماهنگ Precision و Recall که تعادل بین این دو معیار را نشان می‌دهد. | بالاتر از ۷۲٪ |
| **PR-AUC** | توانایی مدل در ایجاد تعادل مناسب بین Precision و Recall در آستانه‌های مختلف پیش‌بینی را نشان می‌دهد. | بالاتر از ۰٫۷۰ |
| **ROC-AUC** | توانایی مدل در تفکیک مشتریان ریزشی از مشتریان غیرریزش‌کننده را اندازه‌گیری می‌کند. | بالاتر از ۰٫۸۰ |
| **Confusion Matrix** | تعداد پیش‌بینی‌های درست و اشتباه مدل را در چهار گروه مثبت واقعی، منفی واقعی، مثبت کاذب و منفی کاذب نمایش می‌دهد. | کاهش مثبت و منفی کاذب |


---

**سوال 9:** موفقیت پروژه از دید مدیر چگونه اندازه‌گیری می‌شود؟

**پاسخ:** اگر کمپین ۵۰۰ نفره سود خالص مثبت داشته باشد و تصمیم قابل‌اجرا و شفافی ارائه شود. مدیر به دقت مدل کاری ندارد، به پول و ریسک تصمیم نگاه می‌کند.

**چرایی پرسش:** مدیران سازمان‌ها معمولاً سؤالات متفاوتی می‌پرسند:

آیا پروژه باعث کاهش ریزش شد؟
آیا سرمایه‌گذاری انجام‌شده ارزش داشت؟ و...
 تعریف موفقیت از دید تصمیم‌گیرنده اصلی.


---

**سوال 10:** اگر سیستم ۹۰٪ دقت داشته باشد اما تیم بازاریابی به آن اعتماد نکند، پروژه موفق است؟ چرا؟

**پاسخ:** خیر. مدلی که استفاده نشود سود نمی‌سازد. ضمناً Accuracy گمراه‌کننده است، 

---

**سوال 11:** مزایای مالی احتمالی پروژه را فهرست کنید.

**پاسخ:** حفظ درآمد مشتریان ارزشمند، کاهش هزینه تماس‌های بی‌اثر و افزایش بازده سرمایه‌گذاری بازاریابی (ROI).

**چرایی پرسش:** 

«مدیر میخواد بدونه در مقابل این هزینه‌ها، چه منفعت مالی به دست خواهیم آورد؟» پس : توجیه اقتصادی پروژه برای مدیر و مالی.

---

**سوال 12:** مزایای احتمالی برای تجربه مشتری را فهرست کنید.

**پاسخ:** پیشنهاد مرتبط‌تر، یادآوری به‌موقع خرید و کاهش پیام‌های بی‌ربط به مشتری.

**چرایی پرسش:** اثر پروژه فقط مالی نیست و رضایت مشتری را هم شامل می‌شود.

---

**سوال 13:** یک Business Objective Statement رسمی بنویسید.

**پاسخ:** «شرکت می‌خواهد با تحلیل تراکنش‌های گذشته، مشتریانی را که در ۹۰ روز آینده خرید معتبری نخواهند داشت پیش از ریزش شناسایی کند و با بودجه تماس برای ۵۰۰ مشتری، سود خالص کمپین را به حداکثر برساند.»

**چرایی پرسش:** معمولا مدیرعامل یا حامی پروژه (Project Sponsor)  فرصت مطالعه ده‌ها صفحه گزارش را ندارد پس این متن رو تحویلش میدیم.. 

---

## بخش ۲: Assess Situation

**سوال 14:** چه اطلاعاتی را قبل از شروع پروژه باید از شرکت دریافت کنیم؟

**پاسخ:** تعریف شرکت از «مشتری فعال و ریزش‌کرده»، حاشیه سود، هزینه هر تماس و تخفیف، سیاست مرجوعی و لغو، دسترسی‌های مجاز به داده و دوره زمانی داده‌های موجود.
> در پروژه های داده کاوی این سوال یکی از واجب ترین و مهم ترین سوالات است، اما در این پروژه ما معنی خاصی ندارد و پروژه به دلیل تمرینی بودن داده ها از قبل در اختیار ما قرار داده شده و این سوال برای کامل بودن document ما است.

**چرایی پرسش:** یک مشاور حرفه‌ای قبل از هر اقدامی، وضعیت موجود را به‌طور کامل بررسی می‌کند.

---

**سوال 15:** حداقل ۱۰ سوال که باید از مدیر شرکت پرسیده شود.

**پاسخ:**

1. از دید شما «مشتری ریزش‌کرده» کیست؟ ۹۰ روز بدون خرید منطقی است؟
2. حفظ کدام گروه از مشتریان مهم‌تر است (پرهزینه یا پرتعداد)؟
3. بودجه کمپین و هزینه هر تماس چقدر است؟
4. حاشیه سود متوسط هر سفارش چقدر است؟
5. حداقل بازده قابل‌قبول کمپین چقدر است؟
6. آیا قبلاً کمپین حفظ مشتری اجرا شده و نتیجه‌اش چه بوده؟
7. آیا امکان اجرای A/B Test وجود دارد؟
8. کدام بازارها اولویت دارند؟
9. مدل هر چند وقت یک‌بار باید به‌روز شود؟
10. محدودیت‌های قانونی ارتباط با مشتری چیست؟

**چرایی پرسش:** پاسخ‌ها فرض‌ها را به واقعیت تبدیل می‌کند (تعریف ریزش، بودجه، سنجش اثر).اگر اطلاعات به‌درستی جمع‌آوری نشوند، حتی بهترین مدل یادگیری ماشین نیز ممکن است مشکل اشتباهی را حل کند.

---

**سوال 16:** حداقل ۱۰ سوال که باید از تیم بازاریابی/CRM پرسیده شود.

**پاسخ:**

1. کدام اقدام (تخفیف، یادآوری، پیشنهاد محصول) تاکنون بهتر جواب داده؟
2. در هر دوره عملاً چند مشتری را می‌توانید پوشش دهید؟
3. لیست در چه قالبی برای شما قابل‌استفاده‌تر است؟
4. کدام مشتریان نباید تماس گرفته شوند؟
5. مشتریان عمده‌فروش و خرده‌فروش چه تفاوتی دارند؟
6. لغو و مرجوعی معمولاً چه دلایلی دارد؟
7. فصل‌ها و مناسبت‌ها چه اثری بر خرید دارند؟
8. چه چیزی باعث می‌شود به خروجی مدل اعتماد کنید؟
9. اطلاعات تماس مشتریان در دسترس است؟
10. نتیجه هر کمپین چگونه ثبت و پیگیری می‌شود؟

---

**سوال 17:** محدودیت‌های مالی پروژه را تحلیل کنید.

**پاسخ:** بودجه تماس فقط برای ۵۰۰ مشتری است، پس مدل باید دقیق‌ترین ۵۰۰ نفر را انتخاب کند. هر تماس بی‌نتیجه هم هزینه است و هم یک جای خالی از ۵۰۰ جا.

**چرایی پرسش:** اگر محدودیت‌ها از ابتدا شناخته نشوند، ممکن است پروژه در میانه راه متوقف شود یا هزینه آن چند برابر شود.

---

**سوال 18:** مهم‌ترین ریسک‌ها را شناسایی و احتمال و شدت اثرشان را مشخص کنید.

**پاسخ:**

| ریسک | احتمال | شدت اثر | اقدام مقابله |
|------|--------|---------|--------------|
| نشت اطلاعات آینده به ویژگی‌ها | متوسط | بسیار زیاد | ویژگی فقط از گذشته، ارزیابی زمانی |
| تعریف ۹۰ روزه، مشتری فصلی را ریزشی نشان دهد | زیاد | زیاد | تحلیل حساسیت آستانه |
| عدم‌توازن کلاس‌ها | متوسط | متوسط | معیارهای Precision/Recall/PR-AUC |
| کیفیت پایین داده (لغو، منفی، گمشده) | زیاد | متوسط | ممیزی مستند، بدون حذف کورکورانه |
| تفسیر علّی اشتباه از اهمیت ویژگی | زیاد | متوسط | تأکید بر همبستگی، نه علیت |
| مدل پیچیده و غیرقابل‌فهم برای بازاریابی | متوسط | متوسط | مدل ساده‌ی قابل‌توضیح کنار مدل پیچیده |

**چرایی پرسش:** قبل از اینکه پروژه آغاز شود، باید از خود بپرسیم:

چه اتفاقاتی ممکن است پروژه را با شکست مواجه کنند؟  
احتمال وقوع هر اتفاق چقدر است؟   
اگر رخ دهد، چه میزان خسارت ایجاد خواهد کرد؟  
آیا می‌توان از وقوع آن جلوگیری کرد؟  


ریسک شناخته شده قابل مدیریت است.


---

**سوال 19:** چه داده‌هایی احتمالاً برای پروژه لازم است؟

**پاسخ:** تراکنش‌های خرید شامل شماره فاکتور، محصول، تعداد، تاریخ، قیمت واحد، شناسه مشتری و کشور، برای حداقل چند ماه متوالی.

**چرایی پرسش:** از ابتدا داده‌های موردنیاز را نشناسیم، ممکن است ماه‌ها روی داده‌های اشتباه کار کنیم.

---

**سوال 20:** اگر داده کافی نباشد چه راهکارهایی پیشنهاد می‌کنید؟

**پاسخ:** ویژگی‌ها را از خود تراکنش‌ها می‌سازیم (RFM، روند خرید، نرخ لغو)، محدودیت‌ها را شفاف گزارش می‌دهیم و از شرکت می‌خواهیم داده‌های بیشتری را از این به بعد ثبت کند.

---

**سوال 21:** اگر داده‌ها خطا داشته باشند چه مشکلاتی ایجاد می‌شود؟

**پاسخ:** لغو یا مقدار منفی اگر به‌اشتباه جمع یا حذف شود، ارزش خرید و فاصله از آخرین خرید را غلط می‌کند. این باعث برچسب ریزش مخدوش و مدل گمراه می‌شود.

---

**سوال 22:** یک SWOT برای پروژه تهیه کنید.

SWOT از چهار بخش تشکیل شده است:

- Strengths (نقاط قوت)
- Weaknesses (نقاط ضعف)
- Opportunities (فرصت‌ها)
- Threats (تهدیدها)



**پاسخ:**

| | |

**سوال 26:** پروژه Classification، Regression، Clustering یا Association Rule است؟

**پاسخ:**

| نوع | وضعیت | دلیل |
|-----|--------|------|
| Classification | ✅ هدف مکمل | خروجی دو‌حالته: ریزش می‌کند یا نه |

**سوال 30:** Recall چیست؟

**پاسخ:** از کل مشتریان واقعاً ریزش‌کننده، چند درصد را پیدا کردیم.
