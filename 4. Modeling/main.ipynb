# Modeling

---

## بخش اول: شناخت انواع مسائل داده‌کاوی

---

## سؤال ۱ — داده‌کاوی را تعریف کنید

فرایند کشف الگوها، روابط و دانش قابل استفاده از داده با روش‌های آماری، محاسباتی و یادگیری ماشین. هدف فقط ساخت مدل نیست؛ اعتبار الگوها و ارزش آن‌ها در تصمیم‌گیری کسب‌وکار نیز اهمیت دارد.

---

## سؤال ۲ — تفاوت Machine Learning و Data Mining چیست؟

Machine Learning روش‌هایی برای یادگیری از داده و پیش‌بینی یا تصمیم‌گیری فراهم می‌کند؛ اما Data Mining فرایند گسترده‌تر کشف دانش است که می‌تواند از یادگیری ماشین، آمار، مصورسازی و قواعد انجمنی استفاده کند.

---

## سؤال ۳ — Supervised Learning چیست؟

یادگیری از نمونه‌هایی که هم ویژگی‌ها و هم برچسب هدف آن‌ها معلوم است. Classification و Regression از مسائل اصلی آن هستند.

**مثال در پروژه:** استفاده از رفتار خرید گذشته مشتریان و برچسب `churn` برای آموزش مدلی که ریزش مشتریان را پیش‌بینی کند.

---

## سؤال ۴ — Unsupervised Learning چیست؟

یادگیری ساختار از داده بدون برچسب هدف؛ نمونه‌های آن شامل Clustering، کاهش ابعاد و کشف ناهنجاری هستند.

**مثال در پروژه:** گروه‌بندی مشتریان بر اساس رفتار خرید، میزان هزینه، تعداد سفارش‌ها و فاصله از آخرین خرید.

---

## سؤال ۵ — Semi-Supervised Learning چیست؟

استفاده هم‌زمان از تعداد کمی نمونه برچسب‌دار و تعداد زیادی نمونه بدون برچسب؛ مناسب زمانی که برچسب‌گذاری پرهزینه است.

**مثال در پروژه:** زمانی که فقط برای بخشی از مشتریان وضعیت ریزش مشخص باشد و بخواهیم از داده‌های بدون برچسب سایر مشتریان نیز استفاده کنیم.

---

## سؤال ۶ — Reinforcement Learning چیست؟

عامل با محیط تعامل می‌کند، اقدام انجام می‌دهد و پاداش می‌گیرد؛ هدف یادگیری سیاستی برای بیشینه‌کردن پاداش تجمعی است.

**مثال در پروژه:** طراحی سیستمی که با بررسی واکنش مشتریان به پیشنهادهای مختلف، به‌تدریج راهبرد مناسب‌تری برای حفظ مشتری انتخاب کند.

---

## سؤال ۷ — چه زمانی از Supervised Learning استفاده می‌کنیم؟

وقتی برچسب معتبر تاریخی وجود دارد و می‌خواهیم برای نمونه‌های جدید خروجی پیش‌بینی کنیم.

**مثال در پروژه:** وقتی بر اساس خریدهای آینده مشتریان، برچسب ریزش ساخته‌ایم و می‌خواهیم ریزش مشتریان جدید را پیش‌بینی کنیم.

---

## سؤال ۸ — چه زمانی از Unsupervised Learning استفاده می‌کنیم؟

وقتی برچسب نداریم یا می‌خواهیم ساختار، گروه‌ها یا نمونه‌های غیرعادی را کشف کنیم.

**مثال در پروژه:** شناسایی گروه‌های مختلف مشتریان مانند مشتریان وفادار، کم‌فعالیت و مشتریان با احتمال بالای ریزش، بدون استفاده از برچسب هدف.

---

## سؤال ۹ — Classification چیست؟

پیش‌بینی یک برچسب گسسته. در مسئله طبقه‌بندی دودویی، خروجی می‌تواند یکی از دو کلاس باشد.

**مثال در پروژه:** پیش‌بینی اینکه مشتری در ۹۰ روز آینده خرید خواهد کرد یا خیر:

- `churn = 1`: مشتری در ۹۰ روز آینده خریدی انجام نمی‌دهد.
- `churn = 0`: مشتری در ۹۰ روز آینده خرید انجام می‌دهد.

---

## سؤال ۱۰ — Regression چیست؟

پیش‌بینی یک مقدار عددی پیوسته؛ معیارهای رایج آن MAE، MSE و RMSE هستند.

**مثال در پروژه:** پیش‌بینی مبلغ خرید آینده مشتری یا میزان هزینه‌ای که احتمالاً در آینده انجام می‌دهد.

---

## سؤال ۱۱ — Clustering چیست؟

گروه‌بندی نمونه‌های مشابه بدون برچسب از پیش تعیین‌شده. کیفیت خوشه‌ها و معنای کسب‌وکاری آن‌ها باید بررسی شود.

**مثال در پروژه:** گروه‌بندی مشتریان بر اساس Recency، Frequency و Monetary برای شناسایی الگوهای رفتاری مختلف.

---

## سؤال ۱۲ — Association Rule Mining چیست؟

کشف هم‌وقوعی‌ها، مانند خرید هم‌زمان دو محصول. معیارهای رایج آن Support، Confidence و Lift هستند؛ هم‌وقوعی به معنی علیت نیست.

**مثال در پروژه:** کشف محصولاتی که مشتریان معمولاً در یک سفارش با یکدیگر خریداری می‌کنند تا از این الگوها برای پیشنهاد محصولات استفاده شود.

---

## سؤال ۱۳ — Anomaly Detection چیست؟

شناسایی نمونه‌هایی که از الگوی معمول فاصله دارند؛ مانند رفتارهای خرید غیرعادی، سفارش‌های مشکوک یا خطاهای ثبت داده. ناهنجاری الزاماً به معنی خطا نیست.

**مثال در پروژه:** شناسایی مشتریانی که به‌صورت غیرمعمول تعداد زیادی سفارش لغوشده یا مقادیر خرید بسیار متفاوت از الگوی معمول دارند.

---

## سؤال ۱۴ — نوع مسئله در مثال‌های مختلف چیست؟

- پیش‌بینی ریزش مشتریان: **Binary Classification**
- پیش‌بینی مبلغ خرید مشتری: **Regression**
- گروه‌بندی مشتریان بر اساس رفتار خرید: **Clustering**
- کشف محصولات پرفروش در کنار یکدیگر: **Association Rule Mining**
- شناسایی رفتارهای خرید غیرعادی: **Anomaly Detection**
- پیشنهاد محصولات مناسب به مشتری: **Recommendation System / Ranking**

---

### سؤال ۱۵ — جدول Problem Type | Example | Algorithms

| Problem Type | Example | Algorithms |
|--------------|---------|------------|
| Classification | ریزش مشتری | Logistic Regression، Decision Tree، Random Forest، Gradient Boosting |
| Regression | پیش‌بینی مبلغ خرید مشتری | Linear Regression، Random Forest Regressor |
| Clustering | تقسیم مشتریان | K-Means |
| Association Rule | پیشنهاد محصول | Apriori |
| Anomaly Detection | شناسایی رفتار خرید غیرعادی | Isolation Forest |

## بخش دوم: تشخیص نوع مسئله فعلی

### سؤال 16 --- مسئله ریزش مشتری از چه نوعی است؟
**Clustering (اصلی)** برای گروه‌بندی رفتاری مشتریان و **Binary Classification (مکمل)** برای پیش‌بینی ریسک ریزش.

**چرایی پرسش:** الگوریتم و معیار درست را از همین نوع انتخاب می‌کنیم.

---

### سؤال 17 --- چرا Clustering روش اصلی است؟
چون مشتریان رفتار بسیار متفاوتی دارند (کم‌خرید، پرخرید، عمده‌خرید) و یک اقدام برای همه مناسب نیست. خوشه‌ها نوع رفتار را نشان می‌دهند و تعیین‌کننده اقدام هر گروه‌اند.

**چرایی پرسش:** توجیه نقش اصلی خوشه‌بندی.

---

### سؤال 18 --- چرا Classification لازم است (مکمل)؟
خوشه‌ها فقط وضعیت فعلی را توصیف می‌کنند و نمی‌گویند چه کسی در آینده ریزش می‌کند. برای رتبه‌بندی ۵۰۰ نفر به احتمال ریزش نیاز داریم.

**چرایی پرسش:** توجیه نقش مکمل پیش‌بینی.

---

### سؤال 19 --- چرا Regression نیست؟
چون خروجی ما مقدار پیوسته نیست و فقط وجود یا نبود خرید در ۹۰ روز آینده را می‌خواهیم. (ارزش مشتری را به‌صورت ویژگی کنار ریسک می‌گیریم.)

**چرایی پرسش:** تفکیک از Regression.

---

### سؤال 20 --- چرا بخش پیش‌بینی Supervised است؟
چون برای هر مشتری برچسب `churn` را از رفتار ۹۰ روز آینده ساخته‌ایم. بخش خوشه‌بندی Unsupervised است و از برچسب استفاده نمی‌کند.

**چرایی پرسش:** تفکیک دو بخش پروژه از نظر نوع یادگیری.

---

### سؤال 21 --- متغیر هدف چیست؟
ستون `churn` (۱ یعنی مشتری در ۹۰ روز پس از تاریخ مرجع هیچ خرید معتبری نداشته است).

**چرایی پرسش:** مدل دقیقاً چه چیزی را پیش‌بینی می‌کند.

---

### سؤال 22 --- چند کلاس وجود دارد؟
دو کلاس: 0 (خرید می‌کند) و 1 (ریزش). نسبتشان را در کد تست می‌بینیم.

**چرایی پرسش:** عدم‌توازن کلاس‌ها را بررسی می‌کنیم.

---

### سؤال 23 --- Binary Classification است یا Multi-Class؟
Binary Classification.

**چرایی پرسش:** معیارهایی مثل PR-AUC و ROC-AUC برای حالت دودویی تعریف می‌شوند.

---

### سؤال 24 --- اگر target حذف شود مسئله به چه نوعی تبدیل می‌شود؟
به Clustering (Unsupervised)؛ یعنی همان بخش اصلی ما.

**چرایی پرسش:** رابطه برچسب و نوع مسئله.

---

### سؤال 25 --- چه معیارهایی برای موفقیت مدل اهمیت دارند؟
خوشه‌بندی: ضریب Silhouette و تفاوت نرخ ریزش بین خوشه‌ها. پیش‌بینی: PR-AUC، Precision@500، Recall، F1، ROC-AUC و Lift. Accuracy به‌تنهایی کافی نیست.

**چرایی پرسش:** مدل با معیار درست برنده می‌شود.

## بخش سوم: Select Modeling Technique

### سؤال 26 --- پنج الگوریتم مناسب برای Binary Classification را معرفی کنید
Logistic Regression، Decision Tree، Random Forest، Gradient Boosting و SVM.

**چرایی پرسش:** گزینه‌ها را می‌شناسیم تا از میانشان آگاهانه انتخاب کنیم.

---

### سؤال 27 --- مزایا و معایب Logistic Regression
مزیت: ساده، سریع، قابل‌توضیح و خروجی احتمالی. عیب: فقط روابط خطی و حساس به مقیاس و همبستگی.

**چرایی پرسش:** مدل ساده و توضیح‌پذیر ما.

---

### سؤال 28 --- مزایا و معایب Decision Tree
مزیت: قوانین قابل‌خواندن و بدون نیاز به مقیاس‌دهی. عیب: ناپایدار و مستعد Overfitting.

**چرایی پرسش:** مدل قابل‌ارائه به بازاریابی.

---

### سؤال 29 --- مزایا و معایب Random Forest
مزیت: دقت و پایداری بالا. عیب: کمتر قابل‌توضیح و کندتر.

**چرایی پرسش:** مدل ترکیبی دقیق‌تر.

---

### سؤال 30 --- مزایا و معایب Gradient Boosting
مزیت: معمولاً بهترین دقت روی داده جدولی. عیب: تنظیم حساس، کندتر و کمتر قابل‌توضیح.

**چرایی پرسش:** پروژه حداقل یک مدل Boosting می‌خواهد.

---

### سؤال 31 --- مزایا و معایب SVM
مزیت: خوب در ابعاد بالا. عیب: کند در داده بزرگ و احتمال خروجی ضعیف، در حالی که ما رتبه‌بندی احتمالی می‌خواهیم.

**چرایی پرسش:** دلیل انتخاب‌نکردن آن.

---

### سؤال 32 --- مزایا و معایب KNN
مزیت: ساده و بدون آموزش. عیب: کند در پیش‌بینی، حساس به مقیاس و ابعاد و بدون مدل قابل‌توضیح.

**چرایی پرسش:** دلیل انتخاب‌نکردن آن.

---

### سؤال 33 --- مزایا و معایب K-Means
مزیت: ساده، سریع و تفسیرپذیر. عیب: فرض خوشه‌های کروی و هم‌اندازه، نیاز به تعیین k و حساسیت به مقیاس و Outlier.

**چرایی پرسش:** K-Means روش اصلی ماست و باید محدودیت‌هایش را بدانیم.

---

### سؤال 34 --- کدام الگوریتم‌ها Explainable هستند؟
Logistic Regression (ضرایب) و Decision Tree (قوانین). Random Forest و Gradient Boosting با Permutation Importance توضیح داده می‌شوند.

**چرایی پرسش:** تیم بازاریابی باید دلیل پیش‌بینی را بفهمد.

---

### سؤال 35 --- Explainability برای تیم بازاریابی چه اهمیتی دارد؟
تیم باید بداند چرا مشتری پرخطر است تا اقدام مناسب بزند و به لیست اعتماد کند.

**چرایی پرسش:** اعتماد کاربر شرط استفاده از مدل است.

---

### سؤال 36 --- اگر Accuracy و Explainability تضاد داشته باشند کدام مهم‌تر است؟
اگر تفاوت دقت ناچیز باشد، مدل توضیح‌پذیر. اگر مدل پیچیده سود بیشتری بسازد، آن را با Permutation Importance همراه می‌کنیم.

**چرایی پرسش:** تصمیم میان مدل ساده و دقیق.

---

### سؤال 37 --- جدول Algorithm | Pros | Cons | Suitable

| Algorithm | Pros | Cons | Suitable |
|-----------|------|------|----------|
| K-Means | ساده، تفسیرپذیر | نیاز به k و مقیاس | ✅ (اصلی) |
| Logistic Regression | ساده، قابل‌توضیح | فقط خطی | ✅ |
| Decision Tree | قوانین قابل‌خواندن | Overfitting | ✅ |
| Random Forest | دقیق و پایدار | کمتر توضیح‌پذیر | ✅ |
| Gradient Boosting | بالاترین دقت | تنظیم حساس | ✅ |
| SVM | ابعاد بالا | کند، احتمال ضعیف | ❌ |
| KNN | ساده | کند، حساس به مقیاس | ❌ |

**چرایی پرسش:** انتخاب الگوریتم یکجا قابل‌مقایسه است.

---

### سؤال 38 --- الگوریتم‌های نهایی خود را انتخاب کنید
خوشه‌بندی: K-Means. پیش‌بینی: Logistic Regression، Decision Tree، Random Forest، Gradient Boosting. خط مبنا: امتیاز ساده RFM.

**چرایی پرسش:** مدل‌هایی که وارد مقایسه می‌شوند.

---

### سؤال 39 --- دلایل انتخاب هر الگوریتم را توضیح دهید
K-Means: شناخت نوع رفتار. Logistic Regression: ساده و توضیح‌پذیر. Decision Tree: قوانین قابل‌خواندن. Random Forest: دقت و پایداری. Gradient Boosting: بهترین دقت احتمالی. RFM: خط مبنا برای سنجش ارزش افزوده.

**چرایی پرسش:** انتخاب‌ها قابل‌دفاع باشند.

---

### سؤال 40 --- پیش‌بینی کنید کدام مدل بهترین عملکرد را خواهد داشت
Gradient Boosting یا Random Forest.

**چرایی پرسش:** پیش‌بینی ما با نتیجه واقعی مقایسه می‌شود.

---

### سؤال 41 --- پیش‌بینی کنید کدام مدل بیشترین Explainability را دارد
Logistic Regression و Decision Tree.

**چرایی پرسش:** مدل قابل‌ارائه به بازاریابی کدام است.

---

### سؤال 42 --- پیش‌بینی کنید کدام مدل بیشترین Overfitting را خواهد داشت
Decision Tree عمیق (بدون محدودیت عمق).

**چرایی پرسش:** می‌دانیم کدام مدل را باید با پارامترهای محدودکننده تنظیم کنیم.

## بخش چهارم: Generate Test Design

### سؤال 43 --- چرا داده را به Train و Test تقسیم می‌کنیم؟
تا عملکرد مدل را روی داده‌ای بسنجیم که ندیده است.

**چرایی پرسش:** ارزیابی صادقانه ممکن شود.

---

### سؤال 44 --- خطر استفاده از کل داده برای آموزش چیست؟
مدل داده را حفظ می‌کند و ارزیابی صادقانه غیرممکن می‌شود.

**چرایی پرسش:** دلیل جداکردن تست.

---

### سؤال 45 --- مفهوم Generalization چیست؟
عملکرد خوب مدل روی داده جدیدی که ندیده است.

**چرایی پرسش:** هدف واقعی مدل همین است.

---

### سؤال 46 --- مفهوم Overfitting چیست؟
مدل آموزش را حفظ می‌کند و روی داده جدید ضعیف است.

**چرایی پرسش:** باید فاصله عملکرد آموزش و تست را ببینیم.

---

### سؤال 47 --- مفهوم Underfitting چیست؟
مدل آن‌قدر ساده است که حتی روی داده آموزش هم ضعیف است.

**چرایی پرسش:** مدل نباید ساده‌تر از حد نیاز باشد.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import hashlib

from sklearn.base import clone
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.cluster import KMeans
from sklearn.inspection import permutation_importance
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
    average_precision_score, confusion_matrix, ConfusionMatrixDisplay,
    roc_curve, roc_auc_score, silhouette_score)

RANDOM_STATE = 42

df = pd.read_csv("churn_prepared_scaled.csv", parse_dates=["ref_date"]) 
raw = pd.read_csv("churn_prepared.csv", parse_dates=["ref_date"])   
assert "churn" in df.columns
assert df["churn"].dropna().isin([0, 1]).all()

X = df.drop(columns=["churn", "CustomerID", "ref_date"]).copy()
y = df["churn"].astype(int).copy()

numeric_features = X.select_dtypes(include=np.number).columns.tolist()
categorical_features = X.select_dtypes(exclude=np.number).columns.tolist()

preprocessor = ColumnTransformer([
    ("num", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]), numeric_features),
    ("cat", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]), categorical_features)
])

def build_pipeline(model):
    return Pipeline([("preprocess", clone(preprocessor)), ("model", model)]) 

print("Rows:", X.shape, "| numeric:", len(numeric_features), "| categorical:", len(categorical_features))
```

### سؤال 48 --- Train/Test Split را پیاده‌سازی کنید
تقسیم ما زمانی است، نه تصادفی: آموزش روی تاریخ مرجع 1 ژوئن (برچسب آن تا 30 اوت می‌رسد) و تست نهایی روی 1 سپتامبر (بدون همپوشانی با برچسب‌های آموزش). ژوئیه و اوت فقط برای بررسی پایداری استفاده می‌شوند.

</div>

```python
train_mask = df["ref_date"] == "2011-06-01"
test_mask = df["ref_date"] == "2011-09-01"

X_train, y_train = X[train_mask], y[train_mask]
X_test, y_test = X[test_mask], y[test_mask]
print("Full:", X.shape, "Train:", X_train.shape, "Test:", X_test.shape)
```

### سؤال 49 --- اگر تقسیم تصادفی 80/20 استفاده کنیم چه می‌شود؟
یک Logistic Regression را یک‌بار با تقسیم تصادفی 80/20 و یک‌بار با تقسیم زمانی آموزش می‌دهیم و PR-AUC را مقایسه می‌کنیم. اگر عدد تصادفی بالاتر باشد، یعنی ارزیابی تصادفی خوش‌بینانه است.

**چرایی پرسش:** شاهد عددی برای اینکه چرا تقسیم زمانی لازم است.

</div>

```python
Xr_train, Xr_test, yr_train, yr_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y)

demo_random = build_pipeline(LogisticRegression(max_iter=3000, random_state=RANDOM_STATE))
demo_random.fit(Xr_train, yr_train)
pr_random = average_precision_score(yr_test, demo_random.predict_proba(Xr_test)[:, 1])

demo_time = build_pipeline(LogisticRegression(max_iter=3000, random_state=RANDOM_STATE))
demo_time.fit(X_train, y_train)
pr_time = average_precision_score(y_test, demo_time.predict_proba(X_test)[:, 1])

print("PR-AUC random 80/20 split:", round(pr_random, 3))
print("PR-AUC temporal split    :", round(pr_time, 3))
```
