# Modeling

---

## بخش اول: شناخت انواع مسائل داده‌کاوی

## سؤال ۱ — داده‌کاوی را تعریف کنید

فرایند کشف الگوها، روابط و دانش قابل استفاده از داده با روش‌های آماری، محاسباتی و یادگیری ماشین. هدف فقط ساخت مدل نیست؛ اعتبار الگوها و ارزش آن‌ها در تصمیم‌گیری کسب‌وکار نیز اهمیت دارد.

## سؤال ۲ — تفاوت Machine Learning و Data Mining چیست؟

Machine Learning روش‌هایی برای یادگیری از داده و پیش‌بینی یا تصمیم‌گیری فراهم می‌کند؛ اما Data Mining فرایند گسترده‌تر کشف دانش است که می‌تواند از یادگیری ماشین، آمار، مصورسازی و قواعد انجمنی استفاده کند.

## سؤال ۳ — Supervised Learning چیست؟

یادگیری از نمونه‌هایی که هم ویژگی‌ها و هم برچسب هدف آن‌ها معلوم است. Classification و Regression از مسائل اصلی آن هستند.

**مثال در پروژه:** استفاده از رفتار خرید گذشته مشتریان و برچسب `churn` برای آموزش مدلی که ریزش مشتریان را پیش‌بینی کند.

## سؤال ۴ — Unsupervised Learning چیست؟

یادگیری ساختار از داده بدون برچسب هدف؛ نمونه‌های آن شامل Clustering، کاهش ابعاد و کشف ناهنجاری هستند.

**مثال در پروژه:** گروه‌بندی مشتریان بر اساس رفتار خرید، میزان هزینه، تعداد سفارش‌ها و فاصله از آخرین خرید.

## سؤال ۵ — Semi-Supervised Learning چیست؟

استفاده هم‌زمان از تعداد کمی نمونه برچسب‌دار و تعداد زیادی نمونه بدون برچسب؛ مناسب زمانی که برچسب‌گذاری پرهزینه است.

**مثال در پروژه:** زمانی که فقط برای بخشی از مشتریان وضعیت ریزش مشخص باشد و بخواهیم از داده‌های بدون برچسب سایر مشتریان نیز استفاده کنیم.

## سؤال ۶ — Reinforcement Learning چیست؟

عامل با محیط تعامل می‌کند، اقدام انجام می‌دهد و پاداش می‌گیرد؛ هدف یادگیری سیاستی برای بیشینه‌کردن پاداش تجمعی است.

**مثال در پروژه:** طراحی سیستمی که با بررسی واکنش مشتریان به پیشنهادهای مختلف، به‌تدریج راهبرد مناسب‌تری برای حفظ مشتری انتخاب کند.

## سؤال ۷ — چه زمانی از Supervised Learning استفاده می‌کنیم؟

وقتی برچسب معتبر تاریخی وجود دارد و می‌خواهیم برای نمونه‌های جدید خروجی پیش‌بینی کنیم.

**مثال در پروژه:** وقتی بر اساس خریدهای آینده مشتریان، برچسب ریزش ساخته‌ایم و می‌خواهیم ریزش مشتریان جدید را پیش‌بینی کنیم.

## سؤال ۸ — چه زمانی از Unsupervised Learning استفاده می‌کنیم؟

وقتی برچسب نداریم یا می‌خواهیم ساختار، گروه‌ها یا نمونه‌های غیرعادی را کشف کنیم.

**مثال در پروژه:** شناسایی گروه‌های مختلف مشتریان مانند مشتریان وفادار، کم‌فعالیت و مشتریان با احتمال بالای ریزش، بدون استفاده از برچسب هدف.

## سؤال ۹ — Classification چیست؟

پیش‌بینی یک برچسب گسسته. در مسئله طبقه‌بندی دودویی، خروجی می‌تواند یکی از دو کلاس باشد.

**مثال در پروژه:** پیش‌بینی اینکه مشتری در ۹۰ روز آینده خرید خواهد کرد یا خیر:

- `churn = 1`: مشتری در ۹۰ روز آینده خریدی انجام نمی‌دهد.
- `churn = 0`: مشتری در ۹۰ روز آینده خرید انجام می‌دهد.

## سؤال ۱۰ — Regression چیست؟

پیش‌بینی یک مقدار عددی پیوسته؛ معیارهای رایج آن MAE، MSE و RMSE هستند.

**مثال در پروژه:** پیش‌بینی مبلغ خرید آینده مشتری یا میزان هزینه‌ای که احتمالاً در آینده انجام می‌دهد.

## سؤال ۱۱ — Clustering چیست؟

گروه‌بندی نمونه‌های مشابه بدون برچسب از پیش تعیین‌شده. کیفیت خوشه‌ها و معنای کسب‌وکاری آن‌ها باید بررسی شود.

**مثال در پروژه:** گروه‌بندی مشتریان بر اساس Recency، Frequency و Monetary برای شناسایی الگوهای رفتاری مختلف.

## سؤال ۱۲ — Association Rule Mining چیست؟

کشف هم‌وقوعی‌ها، مانند خرید هم‌زمان دو محصول. معیارهای رایج آن Support، Confidence و Lift هستند؛ هم‌وقوعی به معنی علیت نیست.

**مثال در پروژه:** کشف محصولاتی که مشتریان معمولاً در یک سفارش با یکدیگر خریداری می‌کنند تا از این الگوها برای پیشنهاد محصولات استفاده شود.

## سؤال ۱۳ — Anomaly Detection چیست؟

شناسایی نمونه‌هایی که از الگوی معمول فاصله دارند؛ مانند رفتارهای خرید غیرعادی، سفارش‌های مشکوک یا خطاهای ثبت داده. ناهنجاری الزاماً به معنی خطا نیست.

**مثال در پروژه:** شناسایی مشتریانی که به‌صورت غیرمعمول تعداد زیادی سفارش لغوشده یا مقادیر خرید بسیار متفاوت از الگوی معمول دارند.

## سؤال ۱۴ — نوع مسئله در مثال‌های مختلف چیست؟

- پیش‌بینی ریزش مشتریان: **Binary Classification**
- پیش‌بینی مبلغ خرید مشتری: **Regression**
- گروه‌بندی مشتریان بر اساس رفتار خرید: **Clustering**
- کشف محصولات پرفروش در کنار یکدیگر: **Association Rule Mining**
- شناسایی رفتارهای خرید غیرعادی: **Anomaly Detection**
- پیشنهاد محصولات مناسب به مشتری: **Recommendation System / Ranking**
